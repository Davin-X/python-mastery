````python
# Q2: Explain Python's Global Interpreter Lock (GIL)
print("The GIL is a mutex that protects access to Python objects, preventing multiple threads from executing Python bytecodes simultaneously.")
print("This makes single-threaded code fast but limits multi-threading performance for CPU-bound tasks.")
print("For CPU-bound work, use multiprocessing; for I/O-bound work, threading works fine.")
print("The GIL is controversial - it simplifies memory management but hinders parallelism.")

# Example
import threading
import time

def cpu_bound_task():
    result = 0
    for i in range(10**7):
        result += i
    return result

print("For CPU-bound tasks: use multiprocessing")
print("For I/O-bound tasks: threading works fine")
````

````python
# Q3: What are decorators and how do they work?
print("Decorators are functions that modify the behavior of other functions or classes.")
print("They wrap functions, allowing you to add functionality before/after execution.")
print("Common uses: logging, authentication, caching, timing.")
print("@decorator syntax is syntactic sugar for function = decorator(function)")

# Example
def timer_decorator(func):
    def wrapper(*args, **kwargs):
        import time
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f'{func.__name__} took {end-start:.2f}s')
        return result
    return wrapper

@timer_decorator
def slow_function():
    import time
    time.sleep(1)
    return 'done'

result = slow_function()  # Prints timing info
````

````python
# Q4: Explain list comprehensions vs map/filter
print("List comprehensions are more readable and often faster than map/filter for simple operations.")
print("List comprehensions: [expression for item in iterable if condition]")
print("map/filter: map(function, iterable), filter(function, iterable)")
print("Use list comprehensions for readability; use map/filter when you don't need a list immediately.")

# Examples
squares_lc = [x**2 for x in range(10) if x % 2 == 0]
squares_map = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, range(10))))

print(f"List comprehension: {squares_lc}")
print(f"Map/filter: {squares_map}")
print("Results are equivalent, but list comprehension is more readable!")

# Nested comprehensions
matrix = [[i*j for j in range(5)] for i in range(5)]
print("Nested comprehensions work too!")
````

````python
# Q1: Explain Big O notation and common complexities
print("Big O notation describes algorithm performance in terms of time/space complexity.")
print("O(1) - Constant: Hash table lookup")
print("O(log n) - Logarithmic: Binary search")
print("O(n) - Linear: Simple loop")
print("O(n log n) - Linearithmic: Merge sort, quicksort")
print("O(n²) - Quadratic: Bubble sort, nested loops")
print("O(2ⁿ) - Exponential: Subset generation")
print("O(n!) - Factorial: Traveling salesman (brute force)")
print("Focus on worst-case, consider average case too.")

# Examples
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print("Binary search is O(log n) - very efficient!")
````

````python
# Q2: Implement a hash table from scratch
print("Hash tables store key-value pairs using hash functions for fast lookups.")
print("Key components: Hash function, Array/buckets, Collision handling")
print("Average O(1) time for get/put operations.")

# Simple hash table implementation
class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.buckets = [[] for _ in range(size)]
    
    def _hash(self, key):
        return hash(key) % self.size
    
    def put(self, key, value):
        index = self._hash(key)
        bucket = self.buckets[index]
        
        for i, (k, v) in enumerate(bucket):
            if k == key:
                bucket[i] = (key, value)
                return
        
        bucket.append((key, value))
    
    def get(self, key):
        index = self._hash(key)
        bucket = self.buckets[index]
        
        for k, v in bucket:
            if k == key:
                return v
        
        raise KeyError(key)

# Usage
ht = HashTable(5)
ht.put('apple', 10)
ht.put('banana', 20)
print(f"apple: {ht.get('apple')}")
print("Hash tables provide average O(1) lookup time!")
````

````python
# Q3: Find the k-th largest element in an array
print("Multiple approaches:")
print("1. Sort array: O(n log n) time")
print("2. Quickselect: O(n) average case")
print("3. Min-heap: O(n log k) time")
print("Quickselect is usually best for finding k-th element.")

import heapq

# Heap method - O(n log k)
def find_kth_largest_heap(nums, k):
    heap = []
    for num in nums:
        if len(heap) < k:
            heapq.heappush(heap, -num)
        elif -num > heap[0]:
            heapq.heappop(heap)
            heapq.heappush(heap, -num)
    return -heap[0]

# Test
nums = [3, 2, 1, 5, 6, 4]
k = 2
result = find_kth_largest_heap(nums, k)
print(f"{k}th largest in {nums} is {result}")
print("Heap method is efficient for small k!")
````

## 📊 Statistics & Probability Interview Questions

Master the statistical concepts that underpin data science and machine learning.

````python
# Q1: Explain p-value and statistical significance
print("P-value is the probability of observing data as extreme as yours, assuming H₀ is true.")
print("Low p-value (< 0.05) suggests data is unlikely under null hypothesis.")
print("Statistical significance doesn't prove practical importance.")
print("Common misconception: p-value is probability that H₀ is true.")
print("Actually: p-value is probability of data given H₀ is true.")

# Example calculation
import numpy as np
import scipy.stats as stats

sample = np.random.normal(36.5, 8, 100)
t_stat, p_value = stats.ttest_1samp(sample, 35)

print(f"Sample mean: {sample.mean():.2f}")
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Statistically significant: {p_value < 0.05}")
````

````python
# Q2: Explain confidence intervals vs hypothesis testing
print("Confidence intervals quantify uncertainty in estimates.")
print("Hypothesis testing assesses whether effects exist.")
print("95% CI means: if we repeated sampling 100 times, 95 intervals would contain true parameter.")
print("CI gives range of plausible values; hypothesis testing gives yes/no decision.")
print("Both related: if CI excludes null value, reject H₀.")

# Confidence interval calculation
data = np.random.normal(100, 15, 50)
mean = np.mean(data)
se = np.std(data, ddof=1) / np.sqrt(len(data))
ci_lower = mean - 1.96 * se
ci_upper = mean + 1.96 * se

print(f"Sample mean: {mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
print("We are 95% confident the true population mean lies in this interval.")
````

````python
# Q3: What's the difference between correlation and causation?
print("Correlation measures relationship between variables; causation implies one causes the other.")
print("Correlation ≠ Causation reasons:")
print("- Third variable (confounding)")
print("- Reverse causation")
print("- Coincidence")
print("- Selection bias")
print("Causation requires: temporal precedence, covariation, elimination of alternatives.")
print("RCTs and proper experimental design help establish causation.")

# Example: Ice cream sales and drowning deaths
temperature = np.random.normal(25, 5, 100)
ice_cream_sales = 50 + 2 * temperature + np.random.normal(0, 10, 100)
drowning_deaths = 5 + 0.1 * temperature + np.random.normal(0, 2, 100)

corr = np.corrcoef(ice_cream_sales, drowning_deaths)[0, 1]
print(f"Correlation between ice cream and drowning: {corr:.3f}")
print("⚠️ Ice cream doesn't cause drowning! Both caused by temperature.")
print("Correlation ≠ Causation!")
````

## 🤖 Machine Learning Interview Questions

Master the ML concepts that are crucial for data science interviews.

````python
# Q1: Explain overfitting and underfitting
print("Overfitting: Model learns noise in training data, performs poorly on new data.")
print("Underfitting: Model too simple to capture underlying patterns.")
print("Overfitting signs: High training accuracy, low validation accuracy.")
print("Underfitting signs: Poor performance on both training and validation.")
print("Solutions - Overfitting: Cross-validation, regularization, early stopping, more data")
print("Solutions - Underfitting: More complex model, better features, reduce regularization")

# Example with polynomial regression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Generate data
X = np.linspace(0, 10, 50)
y = 2 * np.sin(X) + np.random.normal(0, 0.5, 50)

# Try different polynomial degrees
degrees = [1, 4, 15]  # Underfitting, good fit, overfitting

for degree in degrees:
    poly = PolynomialFeatures(degree=degree)
    X_poly = poly.fit_transform(X.reshape(-1, 1))
    
    model = LinearRegression()
    model.fit(X_poly, y)
    
    y_pred = model.predict(X_poly)
    mse = mean_squared_error(y, y_pred)
    
    if degree == 1:
        print(f"Underfitting (degree {degree}): MSE = {mse:.4f}")
    elif degree == 4:
        print(f"Good fit (degree {degree}): MSE = {mse:.4f}")
    else:
        print(f"Overfitting (degree {degree}): MSE = {mse:.4f}")
````

````python
# Q2: What's the difference between precision and recall?
print("Precision: Of predicted positives, how many are actually positive?")
print("Precision = TP / (TP + FP)")
print("Recall: Of actual positives, how many did we correctly identify?")
print("Recall = TP / (TP + FN)")
print("Use precision when: Cost of false positives is high (spam detection)")
print("Use recall when: Cost of false negatives is high (medical diagnosis)")
print("F1-score: Harmonic mean of precision and recall.")

# Example
from sklearn.metrics import precision_score, recall_score, f1_score

# Sample predictions
y_true = [0, 0, 1, 1, 1, 1]
y_pred = [0, 1, 1, 1, 0, 1]

precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f"Precision: {precision:.3f}")
print(f"Recall: {recall:.3f}")
print(f"F1-Score: {f1:.3f}")
print("Precision: 3/4 = 0.75 (3 correct positives out of 4 predicted positives)")
print("Recall: 3/4 = 0.75 (3 correct positives out of 4 actual positives)")
````
