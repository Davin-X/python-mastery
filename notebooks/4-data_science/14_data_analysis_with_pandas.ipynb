## 🧹 Data Cleaning and Preprocessing

Real-world data is messy. Pandas provides powerful tools for cleaning and preprocessing.

````python
# Data cleaning and preprocessing
print("=== DATA CLEANING AND PREPROCESSING ===")

import pandas as pd
import numpy as np

# Create a messy dataset
np.random.seed(42)
messy_data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', None, 'Eve'],
    'age': [25, np.nan, 35, 'thirty', 28, 42],
    'salary': ['$50,000', '$60,000', None, '$70,000', '$55,000', '$80,000'],
    'department': ['HR', 'IT', 'IT', 'Finance', 'HR', np.nan],
    'score': [85.5, 92.3, np.nan, 78.9, 88.1, 95.2]
}

df_messy = pd.DataFrame(messy_data)
print(f"Original messy data:\n{df_messy}")
print(f"\nData types:\n{df_messy.dtypes}")
print(f"\nMissing values:\n{df_messy.isnull().sum()}")

# 1. Handle missing values
print("\n1. HANDLING MISSING VALUES:")
df_clean = df_messy.copy()

# Drop rows with missing names
df_clean = df_clean.dropna(subset=['name'])
print(f"After dropping rows with missing names:\n{df_clean}")

# Fill missing values
df_clean['department'] = df_clean['department'].fillna('Unknown')
df_clean['score'] = df_clean['score'].fillna(df_clean['score'].mean())
print(f"\nAfter filling missing values:\n{df_clean}")

# 2. Data type conversion
print("\n2. DATA TYPE CONVERSION:")

# Convert age to numeric (handle 'thirty')
def convert_age(age):
    if pd.isna(age):
        return np.nan
    if isinstance(age, str):
        age_map = {'thirty': 30}
        return age_map.get(age.lower(), np.nan)
    return age

df_clean['age'] = df_clean['age'].apply(convert_age)
df_clean['age'] = pd.to_numeric(df_clean['age'])

# Clean salary column
df_clean['salary'] = df_clean['salary'].str.replace('$', '').str.replace(',', '').astype(float)

print(f"After type conversion:\n{df_clean}")
print(f"\nData types:\n{df_clean.dtypes}")

# 3. Data validation and cleaning
print("\n3. DATA VALIDATION:")

# Remove duplicates
df_no_duplicates = df_clean.drop_duplicates()
print(f"Duplicates removed: {len(df_clean) - len(df_no_duplicates)} rows")

# Validate data ranges
valid_age = (df_clean['age'] >= 18) & (df_clean['age'] <= 100)
valid_salary = (df_clean['salary'] >= 0) & (df_clean['salary'] <= 500000)
valid_score = (df_clean['score'] >= 0) & (df_clean['score'] <= 100)

df_validated = df_clean[valid_age & valid_salary & valid_score]
print(f"Rows passing validation: {len(df_validated)}/{len(df_clean)}")

# 4. Outlier detection
print("\n4. OUTLIER DETECTION:")

# Using IQR method for outlier detection
def detect_outliers_iqr(data):
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (data < lower_bound) | (data > upper_bound)

# Detect outliers in salary
salary_outliers = detect_outliers_iqr(df_validated['salary'])
print(f"Salary outliers: {salary_outliers.sum()} detected")

# Detect outliers in score
score_outliers = detect_outliers_iqr(df_validated['score'])
print(f"Score outliers: {score_outliers.sum()} detected")

print(f"\nFinal cleaned dataset:\n{df_validated}")
print(f"\nCleaning summary:")
print(f"- Original rows: {len(df_messy)}")
print(f"- Final rows: {len(df_validated)}")
print(f"- Data loss: {((len(df_messy) - len(df_validated)) / len(df_messy) * 100):.1f}%")
````

## 🔍 Data Selection, Filtering, and Transformation

Pandas offers powerful tools for selecting, filtering, and transforming data.

````python
# Data selection, filtering, and transformation
print("=== DATA SELECTION, FILTERING, AND TRANSFORMATION ===")

import pandas as pd
import numpy as np

# Create a comprehensive dataset
np.random.seed(42)
n = 1000

data = {
    'employee_id': range(1, n+1),
    'name': [f'Employee_{i}' for i in range(1, n+1)],
    'department': np.random.choice(['HR', 'IT', 'Finance', 'Marketing', 'Sales'], n),
    'salary': np.random.normal(60000, 15000, n).round(2),
    'experience_years': np.random.randint(1, 25, n),
    'performance_score': np.random.uniform(1, 5, n).round(2),
    'hire_date': pd.date_range('2010-01-01', periods=n, freq='7D')[:n]
}

df = pd.DataFrame(data)

# Ensure salary is positive
df['salary'] = df['salary'].abs()

print(f"Dataset shape: {df.shape}")
print(f"First 5 rows:\n{df.head()}")

# 1. Data Selection
print("\n1. DATA SELECTION:")

# Select columns
salary_dept = df[['name', 'department', 'salary']]
print(f"Selected columns shape: {salary_dept.shape}")

# Select rows by position
first_10 = df.iloc[:10]
print(f"First 10 rows shape: {first_10.shape}")

# Select by label
specific_rows = df.loc[[0, 5, 10, 15]]
print(f"Specific rows shape: {specific_rows.shape}")

# 2. Data Filtering
print("\n2. DATA FILTERING:")

# Filter by single condition
high_performers = df[df['performance_score'] > 4.0]
print(f"High performers (>4.0): {len(high_performers)} employees")

# Filter by multiple conditions
experienced_it = df[(df['department'] == 'IT') & (df['experience_years'] > 10)]
print(f"Experienced IT staff (>10 years): {len(experienced_it)} employees")

# Filter by salary range
mid_range_salary = df[(df['salary'] >= 50000) & (df['salary'] <= 80000)]
print(f"Mid-range salary ($50k-$80k): {len(mid_range_salary)} employees")

# 3. Data Transformation
print("\n3. DATA TRANSFORMATION:")

# Add calculated columns
df['salary_category'] = pd.cut(df['salary'], 
                               bins=[0, 40000, 60000, 80000, float('inf')],
                               labels=['Entry', 'Mid', 'Senior', 'Executive'])

df['experience_level'] = pd.cut(df['experience_years'], 
                                bins=[0, 3, 7, 15, float('inf')],
                                labels=['Junior', 'Mid', 'Senior', 'Expert'])

df['performance_category'] = pd.cut(df['performance_score'], 
                                    bins=[0, 2.5, 3.5, 4.5, 5],
                                    labels=['Poor', 'Average', 'Good', 'Excellent'])

print("Added categorical columns:")
print(f"Salary categories: {df['salary_category'].value_counts().to_dict()}")
print(f"Experience levels: {df['experience_level'].value_counts().to_dict()}")
print(f"Performance categories: {df['performance_category'].value_counts().to_dict()}")

# Transform existing columns
df['annual_bonus'] = (df['salary'] * df['performance_score'] * 0.1).round(2)
df['years_until_retirement'] = 65 - (2024 - df['hire_date'].dt.year)

print(f"\nAdded calculated columns:")
print(f"Annual bonus range: ${df['annual_bonus'].min():.2f} - ${df['annual_bonus'].max():.2f}")
print(f"Years until retirement range: {df['years_until_retirement'].min()} - {df['years_until_retirement'].max()}")

# 4. Advanced Filtering with Query
print("\n4. ADVANCED FILTERING:")

# Using query method
senior_staff = df.query('experience_years > 15 and performance_score > 3.5')
print(f"Senior high-performers: {len(senior_staff)} employees")

# Complex filtering
valuable_employees = df.query('salary > 70000 and performance_score >= 4.0 and department in ["IT", "Finance"]')
print(f"High-value employees in IT/Finance: {len(valuable_employees)} employees")

print(f"\nFinal transformed dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"\nSample transformed data:\n{df[['name', 'salary', 'salary_category', 'annual_bonus']].head()}")
````

## 📈 Grouping, Aggregation, and Pivot Tables

Group data by categories and perform aggregations to gain insights.

````python
# Grouping, aggregation, and pivot tables
print("=== GROUPING, AGGREGATION, AND PIVOT TABLES ===")

import pandas as pd
import numpy as np

# Use the employee dataset from previous section
# (Assuming df is already created with the employee data)

# Create fresh dataset if needed
if 'df' not in locals():
    np.random.seed(42)
    n = 1000
    
    data = {
        'employee_id': range(1, n+1),
        'name': [f'Employee_{i}' for i in range(1, n+1)],
        'department': np.random.choice(['HR', 'IT', 'Finance', 'Marketing', 'Sales'], n),
        'salary': np.random.normal(60000, 15000, n).round(2),
        'experience_years': np.random.randint(1, 25, n),
        'performance_score': np.random.uniform(1, 5, n).round(2),
        'hire_date': pd.date_range('2010-01-01', periods=n, freq='7D')[:n]
    }
    
    df = pd.DataFrame(data)
    df['salary'] = df['salary'].abs()
    
    # Add categorical columns
    df['salary_category'] = pd.cut(df['salary'], 
                                   bins=[0, 40000, 60000, 80000, float('inf')],
                                   labels=['Entry', 'Mid', 'Senior', 'Executive'])
    
    df['experience_level'] = pd.cut(df['experience_years'], 
                                    bins=[0, 3, 7, 15, float('inf')],
                                    labels=['Junior', 'Mid', 'Senior', 'Expert'])

# 1. Basic Grouping and Aggregation
print("1. BASIC GROUPING AND AGGREGATION:")

# Group by department and calculate statistics
dept_stats = df.groupby('department').agg({
    'salary': ['mean', 'min', 'max', 'count'],
    'performance_score': ['mean', 'std'],
    'experience_years': 'mean'
}).round(2)

print("Department Statistics:")
print(dept_stats)

# Group by multiple columns
dept_experience_stats = df.groupby(['department', 'experience_level']).agg({
    'salary': 'mean',
    'performance_score': 'mean',
    'employee_id': 'count'
}).round(2)

print("\nDepartment + Experience Level Statistics:")
print(dept_experience_stats.head(10))

# 2. Advanced Aggregations
print("\n2. ADVANCED AGGREGATIONS:")

# Custom aggregation functions
def salary_range(group):
    return group.max() - group.min()

def top_performer_pct(group):
    return (group >= 4.0).mean() * 100

custom_stats = df.groupby('department').agg({
    'salary': ['mean', salary_range],
    'performance_score': [top_performer_pct, 'max'],
    'experience_years': ['median', 'std']
}).round(2)

print("Custom Aggregation Functions:")
print(custom_stats)

# 3. Pivot Tables
print("\n3. PIVOT TABLES:")

# Basic pivot table
pivot_salary = df.pivot_table(
    values='salary',
    index='department',
    columns='experience_level',
    aggfunc='mean',
    fill_value=0
).round(2)

print("Salary by Department and Experience Level:")
print(pivot_salary)

# Multiple aggregations in pivot table
pivot_multi = df.pivot_table(
    values=['salary', 'performance_score'],
    index='department',
    columns='salary_category',
    aggfunc={'salary': 'mean', 'performance_score': 'count'},
    fill_value=0
).round(2)

print("\nMultiple Metrics Pivot Table:")
print(pivot_multi)

# 4. Cross-tabulations
print("\n4. CROSS-TABULATIONS:")

# Cross-tabulation
dept_performance_xtab = pd.crosstab(
    df['department'], 
    df['performance_category'],
    normalize='index'  # Normalize by row (department)
).round(3)

print("Department vs Performance Category (%):")
print(dept_performance_xtab)

# 5. Time-based Grouping
print("\n5. TIME-BASED GROUPING:")

# Group by year of hire
df['hire_year'] = df['hire_date'].dt.year
yearly_hires = df.groupby('hire_year').size()
print(f"Hires by year:\n{yearly_hires}")

# Group by year and department
year_dept_hires = df.groupby(['hire_year', 'department']).size().unstack(fill_value=0)
print(f"\nHires by year and department:\n{year_dept_hires}")

# Rolling statistics
df_sorted = df.sort_values('hire_date')
rolling_avg_salary = df_sorted.set_index('hire_date')['salary'].rolling(window='365D').mean()
print(f"\nRolling average salary (1 year window):\n{rolling_avg_salary.tail()}")

print("\nGrouping and aggregation operations completed!")
print(f"Dataset provides insights into {len(df)} employees across {df['department'].nunique()} departments")
````
