## 📦 Variables - Storing Data

Variables are containers that store data values. In Python, you don't need to declare variable types explicitly.

````python
# Creating variables of different types
name = "Alice"        # String (text)
age = 25             # Integer (whole number)
height = 5.7         # Float (decimal number)
is_student = True    # Boolean (True/False)

# Printing variables
print(f"My name is {name}")
print(f"I am {age} years old")
print(f"My height is {height} feet")
print(f"Am I a student? {is_student}")
````

## 🔢 Data Types in Python

Python has several built-in data types. Let's explore the most common ones:

````python
# Strings (text data)
greeting = "Hello"
message = 'Welcome to Python!'

# Numbers
whole_number = 42          # Integer
decimal_number = 3.14159   # Float
negative_number = -10      # Negative integer

# Booleans
is_raining = False
is_sunny = True

# Let's check their types
print("Data Types:")
print(f"greeting: {type(greeting)} = {greeting}")
print(f"whole_number: {type(whole_number)} = {whole_number}")
print(f"decimal_number: {type(decimal_number)} = {decimal_number}")
print(f"is_sunny: {type(is_sunny)} = {is_sunny}")
````

## 🧮 Basic Operations

Python supports various operations on different data types.

### Arithmetic Operations

````python
# Arithmetic operations
a = 10
b = 3

print("Arithmetic Operations:")
print(f"a = {a}, b = {b}")
print(f"a + b = {a + b}")    # Addition
print(f"a - b = {a - b}")    # Subtraction
print(f"a * b = {a * b}")    # Multiplication
print(f"a / b = {a / b}")    # Division (float result)
print(f"a // b = {a // b}")  # Floor division (integer result)
print(f"a % b = {a % b}")    # Modulus (remainder)
print(f"a ** b = {a ** b}")  # Exponentiation (power)
````

### String Operations

````python
# String operations
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name

print("String Operations:")
print(f"First name: {first_name}")
print(f"Last name: {last_name}")
print(f"Full name: {full_name}")
print(f"Name length: {len(full_name)}")
print(f"Uppercase: {full_name.upper()}")
print(f"Lowercase: {full_name.lower()}")
print(f"First letter: {full_name[0]}")
print(f"Last letter: {full_name[-1]}")
````

## 🔄 Type Conversion (Casting)

Sometimes you need to convert between different data types.

````python
# Type conversion examples
print("Type Conversion:")

# String to number
number_string = "42"
converted_int = int(number_string)
print(f"String '{number_string}' → int: {converted_int} (type: {type(converted_int)})")

# Number to string
string_number = 42
converted_string = str(string_number)
print(f"Int {string_number} → string: '{converted_string}' (type: {type(converted_string)})")

# Float to int (truncates decimal part)
float_number = 3.9
converted_to_int = int(float_number)
print(f"Float {float_number} → int: {converted_to_int}")

# Int to float
int_number = 42
converted_to_float = float(int_number)
print(f"Int {int_number} → float: {converted_to_float}")
````

## 📊 Constants and Best Practices

Python doesn't have true constants, but we use uppercase names by convention.

````python
# Constants (by convention, uppercase names)
PI = 3.14159
GRAVITY = 9.81
MAX_USERS = 1000
COMPANY_NAME = "Python Mastery"

print("Constants:")
print(f"π (PI): {PI}")
print(f"Gravity: {GRAVITY} m/s²")
print(f"Max users: {MAX_USERS}")
print(f"Company: {COMPANY_NAME}")
````

## 🎭 Advanced Variable Techniques

Python has some powerful variable assignment features.

````python
# Multiple assignment
x, y, z = 1, 2, 3
print(f"Multiple assignment: x={x}, y={y}, z={z}")

# Unpacking
coordinates = (10, 20, 30)
a, b, c = coordinates
print(f"Unpacking tuple: a={a}, b={b}, c={c}")

# Swapping values (Python-style)
left, right = 10, 20
print(f"Before swap: left={left}, right={right}")
left, right = right, left  # Swap in one line!
print(f"After swap: left={left}, right={right}")

# Extended unpacking
numbers = [1, 2, 3, 4, 5]
first, *middle, last = numbers
print(f"Extended unpacking: first={first}, middle={middle}, last={last}")
````

## 🎨 String Formatting

Python has several ways to format strings. F-strings (Python 3.6+) are the most modern and recommended.

````python
# F-strings (modern, recommended)
product = "laptop"
price = 999.99
quantity = 2

print("F-string formatting:")
print(f"Product: {product}")
print(f"Price: ${price}")
print(f"Quantity: {quantity}")
print(f"Total: ${price * quantity:.2f}")
print(f"Product name in uppercase: {product.upper()}")
print(f"Price with commas: {price:,.2f}")

# Old-style formatting (still works)
print("\nOld-style formatting:")
print("Product: %s" % product)
print("Price: $%.2f" % price)
print("Hello %s, you have %d items" % ("Alice", quantity))

# str.format() method
print("\nstr.format() method:")
print("Product: {}".format(product))
print("Price: ${:.2f}".format(price))
print("Hello {name}, you have {count} items".format(name="Alice", count=quantity))
````

## ⚡ Special Characters and Escape Sequences

Special characters help format output and handle special cases.

````python
# Escape sequences
print("Escape Sequences:")
print("Quotes: \"Hello\" and 'World'")
print("New line: First line\nSecond line")
print("Tab:\tIndented text")
print("Backslash: \\")

# Raw strings (ignore escape sequences)
raw_string = r"C:\Users\Alice\Documents\file.txt"
print(f"Raw string: {raw_string}")

# Unicode characters
print("Unicode: π (Pi) and ∑ (Sigma)")
print("Emojis: 😀 📚 🚀")
````

## 📝 Comments and Documentation

Good code is well-documented. Python supports single-line and multi-line comments.

````python
# This is a single-line comment
# Comments are ignored by Python but help other developers understand your code

"""
This is a multi-line comment (docstring)
Docstrings are used to document functions, classes, and modules.
They can be accessed using the __doc__ attribute.
"""

def example_function():
    """This function demonstrates docstrings.
    
    Returns:
        str: A greeting message
    """
    return "Hello from a well-documented function!"

result = example_function()
print(f"Function result: {result}")
print(f"Function docstring: {example_function.__doc__}")
````
