## 📄 Creating and Using Modules

A **module** is a file containing Python code that can be imported and reused in other programs.

````python
# Let's create a simple module
module_content = '''
"""A simple math utilities module."""

def add(a, b):
    """Add two numbers."""
    return a + b

def multiply(a, b):
    """Multiply two numbers."""
    return a * b

def factorial(n):
    """Calculate factorial of n."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)

# Module-level constants
PI = 3.14159
EULER = 2.71828

# Module execution block
if __name__ == "__main__":
    print("Math utils module executed directly")
    print(f"PI: {PI}")
    print(f"5! = {factorial(5)}")
'''

# Write the module to a file
with open('math_utils.py', 'w') as f:
    f.write(module_content)

print("Created math_utils.py module")
````

````python
# Now let's use the module
import math_utils

print("=== USING THE MODULE ===")
print(f"5 + 3 = {math_utils.add(5, 3)}")
print(f"4 × 6 = {math_utils.multiply(4, 6)}")
print(f"PI: {math_utils.PI}")
print(f"EULER: {math_utils.EULER}")
print(f"6! = {math_utils.factorial(6)}")

# Check module attributes
print(f"\nModule name: {math_utils.__name__}")
print(f"Module file: {math_utils.__file__}")
print(f"Module docstring: {math_utils.__doc__}")
````

````python
# Different import styles
print("=== DIFFERENT IMPORT STYLES ===")

# 1. Standard import
import math_utils
result1 = math_utils.add(10, 5)
print(f"1. import math_utils: add(10, 5) = {result1}")

# 2. Import with alias
import math_utils as mu
result2 = mu.multiply(3, 7)
print(f"2. import math_utils as mu: multiply(3, 7) = {result2}")

# 3. From import (specific items)
from math_utils import add, PI
result3 = add(8, 2)
print(f"3. from math_utils import add, PI: add(8, 2) = {result3}, PI = {PI}")

# 4. From import with alias
from math_utils import factorial as fact
result4 = fact(4)
print(f"4. from math_utils import factorial as fact: fact(4) = {result4}")

# 5. Import all (generally not recommended)
from math_utils import *
result5 = multiply(6, 2)
print(f"5. from math_utils import *: multiply(6, 2) = {result5}")

# Check what's imported
import sys
current_module = sys.modules[__name__]
imported_items = [name for name in dir(current_module) if not name.startswith('_')]
print(f"\nImported items in current namespace: {imported_items}")
````

````python
# Module reloading and execution context
print("=== MODULE EXECUTION AND RELOADING ===")

# Run the module directly
print("Running module directly:")
import subprocess
result = subprocess.run([sys.executable, 'math_utils.py'], capture_output=True, text=True)
print(result.stdout)

# Module __name__ attribute
print(f"\nWhen imported: math_utils.__name__ = {math_utils.__name__}")
print(f"When run directly: __name__ = '__main__'")

# Reloading modules (useful during development)
import importlib
print("\nReloading module...")
importlib.reload(math_utils)
print("Module reloaded successfully")

# Module search path
print(f"\nPython module search path:")
for i, path in enumerate(sys.path, 1):
    print(f"  {i}. {path}")
````

## 📦 Creating and Using Packages

A **package** is a directory containing multiple modules and an `__init__.py` file.

````python
# Create a package structure
import os

# Create package directory
os.makedirs('my_package', exist_ok=True)

# Create __init__.py
init_content = '''
"""My Package - A sample Python package."""

__version__ = "1.0.0"
__author__ = "Python Mastery"

# Import key functions to make them available at package level
from .math_utils import add, multiply
from .string_utils import reverse_string, capitalize_words

def get_package_info():
    """Return package information."""
    return {
        "name": "my_package",
        "version": __version__,
        "author": __author__
    }

print("My Package initialized!")
'''

with open('my_package/__init__.py', 'w') as f:
    f.write(init_content)

# Create math_utils.py in the package
math_utils_content = '''
"""Math utilities for the package."""

def add(a, b):
    """Add two numbers."""
    return a + b

def subtract(a, b):
    """Subtract two numbers."""
    return a - b

def power(base, exp):
    """Calculate base to the power of exp."""
    return base ** exp
'''

with open('my_package/math_utils.py', 'w') as f:
    f.write(math_utils_content)

# Create string_utils.py in the package
string_utils_content = '''
"""String utilities for the package."""

def reverse_string(text):
    """Reverse a string."""
    return text[::-1]

def capitalize_words(text):
    """Capitalize first letter of each word."""
    return text.title()

def count_vowels(text):
    """Count vowels in a string."""
    vowels = 'aeiouAEIOU'
    return sum(1 for char in text if char in vowels)
'''

with open('my_package/string_utils.py', 'w') as f:
    f.write(string_utils_content)

# Create a submodule
os.makedirs('my_package/geometry', exist_ok=True)

# Create geometry/__init__.py
with open('my_package/geometry/__init__.py', 'w') as f:
    f.write('# Geometry submodule')

# Create geometry/shapes.py
shapes_content = '''
"""Geometry shapes calculations."""

def circle_area(radius):
    """Calculate circle area."""
    return 3.14159 * radius ** 2

def rectangle_area(width, height):
    """Calculate rectangle area."""
    return width * height
'''

with open('my_package/geometry/shapes.py', 'w') as f:
    f.write(shapes_content)

print("Created package structure: my_package/")
print("Package contents:")
for root, dirs, files in os.walk('my_package'):
    level = root.replace('my_package', '').count(os.sep)
    indent = ' ' * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = ' ' * 2 * (level + 1)
    for file in files:
        print(f"{subindent}{file}")
````

````python
# Using the package
import my_package

print("=== USING THE PACKAGE ===")
print(f"Package info: {my_package.get_package_info()}")

# Using functions imported at package level
print(f"\nUsing package-level imports:")
print(f"add(5, 3) = {my_package.add(5, 3)}")
print(f"multiply(4, 6) = {my_package.multiply(4, 6)}")
print(f"reverse_string('hello') = {my_package.reverse_string('hello')}")
print(f"capitalize_words('hello world') = {my_package.capitalize_words('hello world')}")

# Using submodules directly
print(f"\nUsing submodules directly:")
print(f"my_package.math_utils.power(2, 3) = {my_package.math_utils.power(2, 3)}")
print(f"my_package.string_utils.count_vowels('hello') = {my_package.string_utils.count_vowels('hello')}")

# Using subpackage
from my_package.geometry import shapes
print(f"shapes.circle_area(5) = {shapes.circle_area(5):.2f}")
print(f"shapes.rectangle_area(4, 6) = {shapes.rectangle_area(4, 6)}")
````

## 🛠️ Virtual Environments and Dependency Management

Virtual environments allow you to manage project-specific dependencies.

````python
# Virtual environment management
print("=== VIRTUAL ENVIRONMENT MANAGEMENT ===")

# Check current Python and pip
print(f"Current Python: {sys.executable}")
print(f"Current Python version: {sys.version}")

# Check installed packages
try:
    import pkg_resources
    installed_packages = [f"{pkg.project_name}=={pkg.version}" for pkg in pkg_resources.working_set]
    print(f"\nInstalled packages (first 10): {installed_packages[:10]}")
except ImportError:
    print("\npkg_resources not available")

# Create requirements.txt file
requirements_content = '''
# Requirements for Python Mastery Platform
numpy>=1.21.0
pandas>=1.3.0
matplotlib>=3.4.0
requests>=2.25.0
pytest>=6.2.0
jupyter>=1.0.0
flask>=2.0.0
sqlalchemy>=1.4.0
'''

with open('requirements.txt', 'w') as f:
    f.write(requirements_content)

print("\nCreated requirements.txt")
print("Contents:")
with open('requirements.txt', 'r') as f:
    print(f.read())

# Show virtual environment commands (these would be run in terminal)
print("\nVirtual Environment Commands:")
commands = [
    "# Create virtual environment",
    "python -m venv myenv",
    "",
    "# Activate virtual environment",
    "# On Windows: myenv\\Scripts\\activate",
    "# On macOS/Linux: source myenv/bin/activate",
    "",
    "# Install requirements",
    "pip install -r requirements.txt",
    "",
    "# Check installed packages",
    "pip list",
    "",
    "# Deactivate virtual environment",
    "deactivate"
]

for cmd in commands:
    print(cmd)
````

## 📦 Building Distributable Packages

Learn how to create packages that can be distributed and installed with pip.

````python
# Create setup.py for package distribution
setup_content = '''
"""
Setup script for my_package
"""

from setuptools import setup, find_packages

with open("README.md", "r", encoding="utf-8") as fh:
    long_description = fh.read()

setup(
    name="my_package",
    version="1.0.0",
    author="Python Mastery",
    author_email="info@pythonmastery.com",
    description="A sample Python package for learning",
    long_description=long_description,
    long_description_content_type="text/markdown",
    url="https://github.com/pythonmastery/my_package",
    packages=find_packages(),
    classifiers=[
        "Development Status :: 3 - Alpha",
        "Intended Audience :: Developers",
        "License :: OSI Approved :: MIT License",
        "Operating System :: OS Independent",
        "Programming Language :: Python :: 3",
        "Programming Language :: Python :: 3.8",
        "Programming Language :: Python :: 3.9",
        "Programming Language :: Python :: 3.10",
        "Programming Language :: Python :: 3.11",
    ],
    python_requires=">=3.8",
    install_requires=[
        "numpy>=1.21.0",
    ],
    extras_require={
        "dev": ["pytest>=6.2", "black", "flake8"],
        "web": ["flask>=2.0"],
    },
)
'''

with open('setup.py', 'w') as f:
    f.write(setup_content)

# Create MANIFEST.in for additional files
manifest_content = '''
include README.md
include LICENSE
recursive-include my_package *.txt *.md
global-exclude *.pyc
'''

with open('MANIFEST.in', 'w') as f:
    f.write(manifest_content)

# Create README.md for the package
readme_content = '''
# My Package

A sample Python package demonstrating packaging best practices.

## Installation

```bash
pip install my_package
```

## Usage

```python
import my_package

# Use package functions
result = my_package.add(5, 3)
print(result)  # Output: 8
```

## Development

```bash
# Install development dependencies
pip install -e ".[dev]"

# Run tests
pytest
```
'''

with open('README.md', 'w') as f:
    f.write(readme_content)

print("Created distribution files:")
print("  setup.py - Package configuration")
print("  MANIFEST.in - Additional files to include")
print("  README.md - Package documentation")

# Show package building commands
print("\nPackage Building Commands:")
build_commands = [
    "# Build source distribution",
    "python setup.py sdist",
    "",
    "# Build wheel distribution",
    "python setup.py bdist_wheel",
    "",
    "# Build both",
    "python setup.py sdist bdist_wheel",
    "",
    "# Install locally for testing",
    "pip install -e .",
    "",
    "# Upload to PyPI (requires account)",
    "twine upload dist/*"
]

for cmd in build_commands:
    print(cmd)
````

## 💡 Pro Tips

2. **Import Order**: Standard library → third-party → local imports
3. **Virtual Environments**: Use `python -m venv` for environment creation
4. **Requirements**: Use `pip freeze > requirements.txt` to capture exact versions
5. **Package Testing**: Test your package with `pip install -e .` before publishing
6. **Version Control**: Don't commit `__pycache__`, `*.pyc`, or virtual environment directories
7. **Documentation**: Include comprehensive README and docstrings
8. **Dependencies**: Keep dependencies minimal and well-specified
