## 🧪 Unit Testing with unittest

`unittest` is Python's built-in testing framework, providing a rich set of testing tools.

```python
# Create a module to test
math_operations_content = '''
"""Simple math operations module for testing demonstrations."""

def add(a, b):
    """Add two numbers."""
    return a + b

def subtract(a, b):
    """Subtract two numbers."""
    return a - b

def multiply(a, b):
    """Multiply two numbers."""
    return a * b

def divide(a, b):
    """Divide two numbers."""
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

def factorial(n):
    """Calculate factorial of n."""
    if not isinstance(n, int) or n < 0:
        raise ValueError("Factorial requires non-negative integer")
    if n <= 1:
        return 1
    return n * factorial(n - 1)

def is_prime(n):
    """Check if number is prime."""
    if not isinstance(n, int) or n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

class Calculator:
    """Simple calculator class."""
    
    def __init__(self):
        self.history = []
    
    def add(self, a, b):
        result = a + b
        self.history.append(f"{a} + {b} = {result}")
        return result
    
    def get_history(self):
        return self.history.copy()
'''

# Write the module to test
with open('math_operations.py', 'w') as f:
    f.write(math_operations_content)

print("Created math_operations.py for testing")
```

```python
# Basic unittest example
import unittest
import math_operations

class TestMathOperations(unittest.TestCase):
    """Test cases for math_operations module."""
    
    def test_add(self):
        """Test the add function."""
        self.assertEqual(math_operations.add(2, 3), 5)
        self.assertEqual(math_operations.add(-1, 1), 0)
        self.assertEqual(math_operations.add(0, 0), 0)
    
    def test_subtract(self):
        """Test the subtract function."""
        self.assertEqual(math_operations.subtract(5, 3), 2)
        self.assertEqual(math_operations.subtract(3, 5), -2)
    
    def test_multiply(self):
        """Test the multiply function."""
        self.assertEqual(math_operations.multiply(4, 5), 20)
        self.assertEqual(math_operations.multiply(-2, 3), -6)
        self.assertEqual(math_operations.multiply(0, 10), 0)
    
    def test_divide(self):
        """Test the divide function."""
        self.assertEqual(math_operations.divide(10, 2), 5)
        self.assertEqual(math_operations.divide(7, 2), 3.5)
        
        # Test division by zero
        with self.assertRaises(ValueError):
            math_operations.divide(10, 0)
    
    def test_factorial(self):
        """Test the factorial function."""
        self.assertEqual(math_operations.factorial(0), 1)
        self.assertEqual(math_operations.factorial(1), 1)
        self.assertEqual(math_operations.factorial(5), 120)
        self.assertEqual(math_operations.factorial(10), 3628800)
        
        # Test invalid inputs
        with self.assertRaises(ValueError):
            math_operations.factorial(-1)
        with self.assertRaises(ValueError):
            math_operations.factorial(3.5)
    
    def test_is_prime(self):
        """Test the is_prime function."""
        # Prime numbers
        self.assertTrue(math_operations.is_prime(2))
        self.assertTrue(math_operations.is_prime(3))
        self.assertTrue(math_operations.is_prime(5))
        self.assertTrue(math_operations.is_prime(7))
        self.assertTrue(math_operations.is_prime(11))
        
        # Not prime numbers
        self.assertFalse(math_operations.is_prime(1))
        self.assertFalse(math_operations.is_prime(4))
        self.assertFalse(math_operations.is_prime(6))
        self.assertFalse(math_operations.is_prime(8))
        self.assertFalse(math_operations.is_prime(9))

# Run the tests
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False, verbosity=2)
```

```python
# Test class with fixtures
class TestCalculator(unittest.TestCase):
    """Test cases for Calculator class."""
    
    def setUp(self):
        """Set up test fixtures before each test method."""
        self.calc = math_operations.Calculator()
    
    def tearDown(self):
        """Clean up after each test method."""
        pass
    
    @classmethod
    def setUpClass(cls):
        """Set up class-level fixtures once for all tests."""
        cls.shared_data = "Shared test data"
    
    @classmethod
    def tearDownClass(cls):
        """Clean up class-level fixtures once after all tests."""
        pass
    
    def test_initialization(self):
        """Test calculator initialization."""
        self.assertEqual(self.calc.get_history(), [])
    
    def test_add(self):
        """Test calculator add method."""
        result = self.calc.add(2, 3)
        self.assertEqual(result, 5)
        self.assertEqual(self.calc.get_history(), ["2 + 3 = 5"])
    
    def test_multiple_operations(self):
        """Test multiple calculator operations."""
        self.calc.add(1, 2)
        self.calc.add(3, 4)
        self.calc.add(5, 6)
        
        expected_history = [
            "1 + 2 = 3",
            "3 + 4 = 7",
            "5 + 6 = 11"
        ]
        self.assertEqual(self.calc.get_history(), expected_history)

# Run the calculator tests
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False, verbosity=2)
```

`pytest` is a popular third-party testing framework that makes testing more Pythonic and powerful.

```python
# pytest examples
try:
    import pytest
    PYTEST_AVAILABLE = True
except ImportError:
    PYTEST_AVAILABLE = False
    # Dependency missing? Install once with: pip install -r requirements.txt
    import pytest
    PYTEST_AVAILABLE = True

# Write pytest test file
pytest_content = '''
"""Pytest examples for math_operations module."""

import pytest
import math_operations

# Basic test functions
def test_add():
    """Test the add function."""
    assert math_operations.add(2, 3) == 5
    assert math_operations.add(-1, 1) == 0
    assert math_operations.add(0, 0) == 0

def test_subtract():
    """Test the subtract function."""
    assert math_operations.subtract(5, 3) == 2
    assert math_operations.subtract(3, 5) == -2

def test_multiply():
    """Test the multiply function."""
    assert math_operations.multiply(4, 5) == 20
    assert math_operations.multiply(-2, 3) == -6

def test_divide():
    """Test the divide function."""
    assert math_operations.divide(10, 2) == 5
    assert math_operations.divide(7, 2) == 3.5

def test_divide_by_zero():
    """Test division by zero raises ValueError."""
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        math_operations.divide(10, 0)

# Parametrized tests
@pytest.mark.parametrize("a,b,expected", [
    (1, 1, 1),
    (2, 1, 2),
    (3, 1, 6),
    (4, 1, 24),
    (5, 1, 120),
])
def test_factorial(a, b, expected):
    """Test factorial with parametrized inputs."""
    assert math_operations.factorial(a) == expected

# Test with invalid inputs
@pytest.mark.parametrize("input_value", [
    -1,
    -5,
    3.5,
    "not_a_number",
])
def test_factorial_invalid_inputs(input_value):
    """Test factorial with invalid inputs."""
    with pytest.raises(ValueError):
        math_operations.factorial(input_value)

# Fixtures
@pytest.fixture
def calculator():
    """Create a calculator instance for testing."""
    return math_operations.Calculator()

def test_calculator_initialization(calculator):
    """Test calculator initialization."""
    assert calculator.get_history() == []

def test_calculator_add(calculator):
    """Test calculator add method."""
    result = calculator.add(2, 3)
    assert result == 5
    assert calculator.get_history() == ["2 + 3 = 5"]

def test_calculator_multiple_operations(calculator):
    """Test multiple calculator operations."""
    calculator.add(1, 2)
    calculator.add(3, 4)
    calculator.add(5, 6)
    
    expected_history = [
        "1 + 2 = 3",
        "3 + 4 = 7",
        "5 + 6 = 11"
    ]
    assert calculator.get_history() == expected_history

# Prime number tests
@pytest.mark.parametrize("number,expected", [
    (1, False),
    (2, True),
    (3, True),
    (4, False),
    (5, True),
    (6, False),
    (7, True),
    (8, False),
    (9, False),
    (11, True),
    (13, True),
    (15, False),
])
def test_is_prime(number, expected):
    """Test is_prime function with various inputs."""
    assert math_operations.is_prime(number) == expected
'''

# Write pytest file
with open('test_math_operations.py', 'w') as f:
    f.write(pytest_content)

print("Created test_math_operations.py with pytest examples")
```

## 🎭 Mocking and Test Isolation

Mocking allows you to isolate code under test by replacing dependencies with controlled substitutes.

```python
# Create a module that uses external dependencies
api_client_content = '''
"""API client module for testing mocking demonstrations."""

import requests
import time

class APIClient:
    """Simple API client for demonstration."""
    
    def __init__(self, base_url):
        self.base_url = base_url
    
    def get_user(self, user_id):
        """Get user data from API."""
        url = f"{self.base_url}/users/{user_id}"
        response = requests.get(url)
        if response.status_code == 200:
            return response.json()
        else:
            raise Exception(f"API request failed: {response.status_code}")
    
    def get_current_time(self):
        """Get current time (for testing time-dependent code)."""
        return time.time()

def process_user_data(api_client, user_id):
    """Process user data from API."""
    user_data = api_client.get_user(user_id)
    
    # Process the data
    if user_data:
        return {
            "id": user_data.get("id"),
            "name": user_data.get("name", "").upper(),
            "email": user_data.get("email", ""),
            "processed_at": api_client.get_current_time()
        }
    return None
'''

# Write the API client module
with open('api_client.py', 'w') as f:
    f.write(api_client_content)

print("Created api_client.py for mocking demonstrations")
```

```python
# Mocking examples
import unittest.mock as mock
import api_client

# unittest.mock examples
class TestAPIClient(unittest.TestCase):
    """Test API client with mocking."""
    
    def setUp(self):
        self.client = api_client.APIClient("https://api.example.com")
    
    @mock.patch('api_client.requests.get')
    def test_get_user_success(self, mock_get):
        """Test successful user retrieval with mocked API."""
        # Mock the response
        mock_response = mock.Mock()
        mock_response.status_code = 200
        mock_response.json.return_value = {
            "id": 123,
            "name": "John Doe",
            "email": "john@example.com"
        }
        mock_get.return_value = mock_response
        
        # Test the method
        result = self.client.get_user(123)
        
        # Assertions
        self.assertEqual(result["id"], 123)
        self.assertEqual(result["name"], "John Doe")
        mock_get.assert_called_once_with("https://api.example.com/users/123")
    
    @mock.patch('api_client.requests.get')
    def test_get_user_failure(self, mock_get):
        """Test user retrieval failure with mocked API."""
        # Mock the response
        mock_response = mock.Mock()
        mock_response.status_code = 404
        mock_get.return_value = mock_response
        
        # Test the method
        with self.assertRaises(Exception):
            self.client.get_user(999)
    
    @mock.patch('api_client.time.time')
    def test_get_current_time(self, mock_time):
        """Test time-dependent method with mocked time."""
        mock_time.return_value = 1234567890.0
        
        result = self.client.get_current_time()
        
        self.assertEqual(result, 1234567890.0)
        mock_time.assert_called_once()

# Test the mocking examples
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False, verbosity=2)
```

## 🔄 Test-Driven Development (TDD)

TDD is a development approach where you write tests first, then implement the code to make the tests pass.

```python
# TDD Example: String Processor
print("=== TEST-DRIVEN DEVELOPMENT EXAMPLE ===")

# First, write the tests (Red phase)
tdd_tests = '''
"""TDD tests for StringProcessor class."""

import pytest

# Import the class we'll create
from string_processor import StringProcessor

class TestStringProcessor:
    """Test cases for StringProcessor using TDD approach."""
    
    def test_reverse_string(self):
        """Test string reversal."""
        processor = StringProcessor()
        assert processor.reverse_string("hello") == "olleh"
        assert processor.reverse_string("Python") == "nohtyP"
        assert processor.reverse_string("") == ""
    
    def test_capitalize_words(self):
        """Test word capitalization."""
        processor = StringProcessor()
        assert processor.capitalize_words("hello world") == "Hello World"
        assert processor.capitalize_words("python programming") == "Python Programming"
    
    def test_count_vowels(self):
        """Test vowel counting."""
        processor = StringProcessor()
        assert processor.count_vowels("hello") == 2  # e, o
        assert processor.count_vowels("AEIOU") == 5
        assert processor.count_vowels("bcdfg") == 0
        assert processor.count_vowels("") == 0
    
    def test_is_palindrome(self):
        """Test palindrome checking."""
        processor = StringProcessor()
        assert processor.is_palindrome("radar") == True
        assert processor.is_palindrome("hello") == False
        assert processor.is_palindrome("A man a plan a canal Panama") == True
        assert processor.is_palindrome("") == True
        assert processor.is_palindrome("a") == True
    
    def test_reverse_string_invalid_input(self):
        """Test reverse_string with invalid inputs."""
        processor = StringProcessor()
        with pytest.raises(TypeError):
            processor.reverse_string(123)
        with pytest.raises(TypeError):
            processor.reverse_string(None)
'''

# Write the TDD test file
with open('test_string_processor.py', 'w') as f:
    f.write(tdd_tests)

# Now implement the class to make tests pass (Green phase)
string_processor_content = '''
"""StringProcessor class - built using TDD approach."""

class StringProcessor:
    """A class for processing strings with various operations."""
    
    def reverse_string(self, text):
        """Reverse a string."""
        if not isinstance(text, str):
            raise TypeError("Input must be a string")
        return text[::-1]
    
    def capitalize_words(self, text):
        """Capitalize first letter of each word."""
        if not isinstance(text, str):
            raise TypeError("Input must be a string")
        return text.title()
    
    def count_vowels(self, text):
        """Count vowels in a string."""
        if not isinstance(text, str):
            raise TypeError("Input must be a string")
        vowels = 'aeiouAEIOU'
        return sum(1 for char in text if char in vowels)
    
    def is_palindrome(self, text):
        """Check if string is a palindrome."""
        if not isinstance(text, str):
            raise TypeError("Input must be a string")
        cleaned = ''.join(c.lower() for c in text if c.isalnum())
        return cleaned == cleaned[::-1]
'''

# Write the implementation
with open('string_processor.py', 'w') as f:
    f.write(string_processor_content)

print("Created TDD implementation: string_processor.py and test_string_processor.py")
print("\nTDD Process:")
print("1. 🔴 RED: Write failing tests first")
print("2. 🟢 GREEN: Implement minimal code to pass tests")
print("3. 🔵 REFACTOR: Clean up code while keeping tests passing")
```

```python
# Run the TDD tests
if PYTEST_AVAILABLE:
    print("\n=== RUNNING TDD TESTS ===")
    import subprocess
    import sys
    
    result = subprocess.run([sys.executable, '-m', 'pytest', 'test_string_processor.py', '-v'], 
                          capture_output=True, text=True)
    
    print("TDD Test Results:")
    print(result.stdout)
    if result.stderr:
        print("Errors:")
        print(result.stderr)
else:
    print("pytest not available for TDD demonstration")
```

## 🚀 Continuous Integration and Testing Automation

Learn how to automate testing in CI/CD pipelines.

```python
# CI/CD configuration examples
print("=== CONTINUOUS INTEGRATION CONFIGURATIONS ===")

# GitHub Actions workflow
github_actions = '''
# .github/workflows/test.yml
name: Test

on: [push, pull_request]

jobs:
  test:
    runs-on: ubuntu-latest
    strategy:
      matrix:
        python-version: ["3.8", "3.9", "3.10", "3.11"]
    
    steps:
    - uses: actions/checkout@v3
    - name: Set up Python ${{ matrix.python-version }}
      uses: actions/setup-python@v4
      with:
        python-version: ${{ matrix.python-version }}
    - name: Install dependencies
      run: |
        python -m pip install --upgrade pip
        pip install -r requirements.txt
        pip install pytest coverage
    - name: Run tests
      run: pytest --cov=src --cov-report=xml
    - name: Upload coverage to Codecov
      uses: codecov/codecov-action@v3
'''

# Write GitHub Actions config
os.makedirs('.github/workflows', exist_ok=True)
with open('.github/workflows/test.yml', 'w') as f:
    f.write(github_actions)

# pytest configuration
pytest_ini = '''
# pytest.ini
[tool:pytest]
testpaths = tests
python_files = test_*.py
python_classes = Test*
python_functions = test_*
addopts = -v --cov=src --cov-report=term-missing
markers =
    slow: marks tests as slow (deselect with '-m "not slow"')
    integration: marks tests as integration tests
'''

# Write pytest config
with open('pytest.ini', 'w') as f:
    f.write(pytest_ini)

# tox configuration for multiple Python versions
tox_ini = '''
# tox.ini
[tox]
envlist = py38, py39, py310, py311

[testenv]
deps =
    pytest
    coverage
commands =
    pytest --cov=src
'''

# Write tox config
with open('tox.ini', 'w') as f:
    f.write(tox_ini)

print("Created CI/CD configuration files:")
print("  .github/workflows/test.yml - GitHub Actions")
print("  pytest.ini - pytest configuration")
print("  tox.ini - Multi-version testing")

# Show commands for different CI systems
ci_commands = {
    "GitHub Actions": "Automatically runs on push/PR",
    "Local pytest": "pytest",
    "With coverage": "pytest --cov=src",
    "Multi-version": "tox",
    "CI-specific": "Check .github/workflows/test.yml"
}

print("\nTesting Commands:")
for system, command in ci_commands.items():
    print(f"  {system}: {command}")
```

5. **Test Edge Cases**: Don't just test happy paths
