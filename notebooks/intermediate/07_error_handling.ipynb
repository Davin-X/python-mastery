## 🚨 Understanding Exceptions

Exceptions are Python's way of handling errors and unexpected situations during program execution.

```python
# Common built-in exceptions
print("=== COMMON BUILT-IN EXCEPTIONS ===")

# ValueError - inappropriate value
try:
    int("not_a_number")
except ValueError as e:
    print(f"ValueError: {e}")

# TypeError - operation on inappropriate type
try:
    "string" + 5
except TypeError as e:
    print(f"TypeError: {e}")

# ZeroDivisionError - division by zero
try:
    10 / 0
except ZeroDivisionError as e:
    print(f"ZeroDivisionError: {e}")

# IndexError - sequence index out of range
try:
    my_list = [1, 2, 3]
    my_list[10]
except IndexError as e:
    print(f"IndexError: {e}")

# KeyError - dictionary key not found
try:
    my_dict = {"a": 1, "b": 2}
    my_dict["c"]
except KeyError as e:
    print(f"KeyError: {e}")

# FileNotFoundError - file not found
try:
    open("nonexistent_file.txt")
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")

# AttributeError - attribute not found
try:
    my_list.nonexistent_method()
except AttributeError as e:
    print(f"AttributeError: {e}")
```

## 🛡️ Try/Except/Else/Finally Blocks

Python's exception handling uses try/except blocks with optional else and finally clauses.

```python
# Basic try/except
print("=== BASIC TRY/EXCEPT ===")

def divide_numbers(a, b):
    """Divide two numbers with basic error handling."""
    try:
        result = a / b
        print(f"{a} ÷ {b} = {result}")
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None

# Test cases
divide_numbers(10, 2)   # Normal case
divide_numbers(10, 0)   # Error case
divide_numbers(10, 5)   # Another normal case
```

```python
# Multiple exception types
print("\n=== MULTIPLE EXCEPTION TYPES ===")

def process_data(data):
    """Process different types of data with comprehensive error handling."""
    try:
        # Try to convert to number
        if isinstance(data, str):
            number = float(data)
        else:
            number = float(data)
        
        # Calculate result
        result = number ** 2 + 10
        print(f"Processed {data} → {number} → {result}")
        return result
        
    except ValueError as e:
        print(f"ValueError: Cannot convert '{data}' to number - {e}")
        return None
        
    except TypeError as e:
        print(f"TypeError: Invalid type for processing - {e}")
        return None
        
    except OverflowError as e:
        print(f"OverflowError: Number too large - {e}")
        return None
        
    except Exception as e:
        print(f"Unexpected error: {type(e).__name__} - {e}")
        return None

# Test with various inputs
test_data = [5, "3.14", "not_a_number", None, [1, 2, 3], {"key": "value"}]
for data in test_data:
    process_data(data)
```

```python
# Complete try/except/else/finally
print("\n=== COMPLETE ERROR HANDLING STRUCTURE ===")

def read_file_safely(filename):
    """Read a file with complete error handling."""
    file = None
    try:
        # Try to open file
        file = open(filename, 'r', encoding='utf-8')
        print(f"Successfully opened: {filename}")
        
        # Try to read content
        content = file.read()
        word_count = len(content.split())
        
        print(f"Read {len(content)} characters, {word_count} words")
        return content
        
    except FileNotFoundError:
        print(f"File not found: {filename}")
        return None
        
    except PermissionError:
        print(f"Permission denied: {filename}")
        return None
        
    except UnicodeDecodeError as e:
        print(f"Encoding error in {filename}: {e}")
        return None
        
    except Exception as e:
        print(f"Unexpected error reading {filename}: {e}")
        return None
        
    else:
        # Executed only if no exception occurred
        print(f"File processing completed successfully for {filename}")
        
    finally:
        # Always executed, regardless of exceptions
        if file:
            file.close()
            print(f"File closed: {filename}")
        print(f"Finished processing request for {filename}")

# Create a test file
with open('test_file.txt', 'w') as f:
    f.write("This is a test file.\nIt contains some sample text.\nFor testing file operations.")

# Test the function
print("Testing read_file_safely:")
content = read_file_safely('test_file.txt')  # Should succeed
print("-" * 50)
content = read_file_safely('nonexistent.txt')  # Should fail
print("-" * 50)
content = read_file_safely('test_file.txt')  # Should succeed again
```

## 🚀 Raising Custom Exceptions

You can create and raise your own exceptions for specific error conditions in your applications.

```python
# Custom exceptions
print("=== CUSTOM EXCEPTIONS ===")

class ValidationError(Exception):
    """Raised when data validation fails."""
    pass

class InsufficientFundsError(Exception):
    """Raised when account has insufficient funds."""
    
    def __init__(self, balance, amount, message="Insufficient funds"):
        self.balance = balance
        self.amount = amount
        self.message = message
        super().__init__(f"{message}: balance=${balance:.2f}, needed=${amount:.2f}")

class InvalidAgeError(ValueError):
    """Raised when age is invalid."""
    
    def __init__(self, age, message="Age must be between 0 and 150"):
        self.age = age
        super().__init__(f"{message}: got {age}")

# Example usage
def validate_age(age):
    """Validate age with custom exception."""
    if not isinstance(age, (int, float)):
        raise ValidationError(f"Age must be a number, got {type(age)}")
    if age < 0:
        raise InvalidAgeError(age, "Age cannot be negative")
    if age > 150:
        raise InvalidAgeError(age)
    return True

# Test custom exceptions
test_ages = [25, -5, 200, "twenty", 75]

for age in test_ages:
    try:
        validate_age(age)
        print(f"✓ Age {age} is valid")
    except ValidationError as e:
        print(f"✗ ValidationError: {e}")
    except InvalidAgeError as e:
        print(f"✗ InvalidAgeError: {e}")
    except Exception as e:
        print(f"✗ Unexpected error: {e}")
```

```python
# Bank account with custom exceptions
print("\n=== BANK ACCOUNT WITH CUSTOM EXCEPTIONS ===")

class BankAccount:
    """Bank account with custom exception handling."""
    
    def __init__(self, account_number, owner, initial_balance=0):
        if initial_balance < 0:
            raise ValidationError("Initial balance cannot be negative")
        
        self.account_number = account_number
        self.owner = owner
        self._balance = initial_balance
        self._transaction_history = []
    
    @property
    def balance(self):
        """Get current balance."""
        return self._balance
    
    def deposit(self, amount):
        """Deposit money."""
        if amount <= 0:
            raise ValidationError(f"Deposit amount must be positive, got {amount}")
        
        self._balance += amount
        self._transaction_history.append(f"Deposit: +${amount:.2f}")
        return self._balance
    
    def withdraw(self, amount):
        """Withdraw money."""
        if amount <= 0:
            raise ValidationError(f"Withdrawal amount must be positive, got {amount}")
        
        if amount > self._balance:
            raise InsufficientFundsError(self._balance, amount)
        
        self._balance -= amount
        self._transaction_history.append(f"Withdrawal: -${amount:.2f}")
        return self._balance
    
    def transfer(self, target_account, amount):
        """Transfer money to another account."""
        # Withdraw from this account
        self.withdraw(amount)
        
        # Deposit to target account
        try:
            target_account.deposit(amount)
        except Exception:
            # Rollback the withdrawal if deposit fails
            self.deposit(amount)
            raise

# Test the bank account
try:
    account1 = BankAccount("ACC001", "Alice", 1000)
    account2 = BankAccount("ACC002", "Bob", 500)
    
    print(f"Account 1 balance: ${account1.balance}")
    print(f"Account 2 balance: ${account2.balance}")
    
    # Valid operations
    account1.deposit(200)
    account1.withdraw(150)
    account1.transfer(account2, 100)
    
    print(f"After operations:")
    print(f"Account 1 balance: ${account1.balance}")
    print(f"Account 2 balance: ${account2.balance}")
    
except (ValidationError, InsufficientFundsError) as e:
    print(f"Account operation failed: {e}")
except Exception as e:
    print(f"Unexpected error: {e}")

# Test error cases
print("\nTesting error cases:")
try:
    account1.withdraw(2000)  # Insufficient funds
except InsufficientFundsError as e:
    print(f"Insufficient funds: {e}")

try:
    account1.deposit(-100)  # Invalid amount
except ValidationError as e:
    print(f"Validation error: {e}")
```

## 🔍 Debugging Techniques

Debugging is the process of finding and fixing errors in your code. Python provides several tools for effective debugging.

```python
# Basic debugging with print statements
print("=== BASIC DEBUGGING ===")

def calculate_average(numbers):
    """Calculate average with debugging prints."""
    print(f"DEBUG: Input numbers: {numbers}")
    
    if not numbers:
        print("DEBUG: Empty list provided, returning 0")
        return 0
    
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    
    print(f"DEBUG: Total = {total}, Count = {count}, Average = {average}")
    return average

# Test with debugging
result1 = calculate_average([1, 2, 3, 4, 5])
print(f"Result: {result1}\n")

result2 = calculate_average([])
print(f"Result: {result2}")
```

```python
# Using assertions for debugging
print("\n=== ASSERTIONS FOR DEBUGGING ===")

def divide_and_check(a, b):
    """Divide with assertions for debugging."""
    # Pre-conditions
    assert isinstance(a, (int, float)), f"'a' must be a number, got {type(a)}"
    assert isinstance(b, (int, float)), f"'b' must be a number, got {type(b)}"
    assert b != 0, "Cannot divide by zero"
    
    result = a / b
    
    # Post-conditions
    assert isinstance(result, (int, float)), "Result should be a number"
    
    return result

# Test assertions
try:
    result = divide_and_check(10, 2)
    print(f"10 ÷ 2 = {result}")
except AssertionError as e:
    print(f"Assertion failed: {e}")

# This will trigger an assertion
try:
    result = divide_and_check(10, 0)
    print(f"Result: {result}")
except AssertionError as e:
    print(f"Assertion failed: {e}")

# Disable assertions in production (python -O script.py)
print("\nNote: Assertions can be disabled with python -O for production")
```

## 📝 Logging for Application Monitoring

Logging is essential for monitoring application behavior, debugging issues, and maintaining production systems.

```python
# Basic logging setup
print("=== BASIC LOGGING ===")
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# Create logger
logger = logging.getLogger(__name__)

# Log messages at different levels
logger.debug("This is a debug message")
logger.info("This is an info message")
logger.warning("This is a warning message")
logger.error("This is an error message")
logger.critical("This is a critical message")

print("\nLogging levels (in order of severity):")
levels = [
    ("DEBUG", "Detailed information for debugging"),
    ("INFO", "General information about program execution"),
    ("WARNING", "Something unexpected happened, but program continues"),
    ("ERROR", "A serious problem that prevents some functionality"),
    ("CRITICAL", "A very serious error that may cause program termination")
]

for level, description in levels:
    print(f"  {level:<8}: {description}")
```

```python
# Advanced logging configuration
print("\n=== ADVANCED LOGGING CONFIGURATION ===")

# Create custom logger with file output
app_logger = logging.getLogger('my_app')
app_logger.setLevel(logging.DEBUG)

# Create formatters
file_formatter = logging.Formatter(
    '%(asctime)s - %(name)s - %(levelname)s - %(funcName)s:%(lineno)d - %(message)s'
)
console_formatter = logging.Formatter(
    '%(levelname)s: %(message)s'
)

# Create handlers
file_handler = logging.FileHandler('app.log')
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(file_formatter)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(console_formatter)

# Add handlers to logger
app_logger.addHandler(file_handler)
app_logger.addHandler(console_handler)

# Test the configured logger
def process_user_data(user_id, data):
    """Process user data with logging."""
    app_logger.info(f"Processing data for user {user_id}")
    
    try:
        if not isinstance(data, dict):
            raise ValueError("Data must be a dictionary")
        
        if 'name' not in data:
            app_logger.warning(f"User {user_id} data missing 'name' field")
            data['name'] = 'Unknown'
        
        result = {
            'user_id': user_id,
            'processed_name': data['name'].upper(),
            'data_length': len(str(data))
        }
        
        app_logger.debug(f"Processed data: {result}")
        app_logger.info(f"Successfully processed user {user_id}")
        
        return result
        
    except Exception as e:
        app_logger.error(f"Failed to process user {user_id}: {e}")
        raise

# Test logging in action
test_users = [
    (1, {'name': 'Alice', 'age': 25}),
    (2, {'age': 30}),  # Missing name
    (3, "invalid_data")  # Invalid format
]

for user_id, data in test_users:
    try:
        result = process_user_data(user_id, data)
        print(f"✓ Processed user {user_id}")
    except Exception as e:
        print(f"✗ Failed to process user {user_id}: {e}")

print("\nCheck 'app.log' for detailed log output")
# Display log file contents
try:
    with open('app.log', 'r') as log_file:
        log_content = log_file.read()
        print("Recent log entries:")
        # Show last few lines
        lines = log_content.strip().split('\n')
        for line in lines[-5:]:
            print(f"  {line}")
except FileNotFoundError:
    print("Log file not found")
```

## 🛡️ Writing Defensive Code

Defensive programming involves anticipating potential errors and handling them gracefully.

```python
# Defensive programming examples
print("=== DEFENSIVE PROGRAMMING ===")

def safe_divide(a, b, default=0):
    """Safely divide two numbers with defensive checks."""
    # Type checking
    if not isinstance(a, (int, float)):
        raise TypeError(f"Numerator must be a number, got {type(a)}")
    if not isinstance(b, (int, float)):
        raise TypeError(f"Denominator must be a number, got {type(b)}")
    
    # Value checking
    if abs(a) > 1e10:
        raise ValueError("Numerator too large")
    if b == 0:
        return default
    
    return a / b

def process_list_safely(data_list, operation):
    """Process a list with defensive programming."""
    if not isinstance(data_list, list):
        raise TypeError("Input must be a list")
    
    if not data_list:
        return []
    
    results = []
    errors = []
    
    for i, item in enumerate(data_list):
        try:
            result = operation(item)
            results.append(result)
        except Exception as e:
            errors.append(f"Item {i}: {e}")
            results.append(None)  # Placeholder for failed items
    
    return {
        'results': results,
        'errors': errors,
        'success_rate': len([r for r in results if r is not None]) / len(results)
    }

# Test defensive functions
print("Testing safe_divide:")
test_cases = [(10, 2), (10, 0), (10, "2"), ("10", 2)]
for a, b in test_cases:
    try:
        result = safe_divide(a, b)
        print(f"{a} ÷ {b} = {result}")
    except Exception as e:
        print(f"{a} ÷ {b} failed: {e}")

print("\nTesting process_list_safely:")
data = [2, 4, 0, 6, "invalid", 8]
result = process_list_safely(data, lambda x: 10 / x)
print(f"Results: {result['results']}")
print(f"Errors: {result['errors']}")
print(f"Success rate: {result['success_rate']:.1%}")
```
