## 📖 Basic File Operations

File operations in Python involve opening, reading/writing, and closing files. Always use context managers for safe file handling.

```python
# Creating a sample file for demonstration
sample_content = """Hello, World!
This is a sample text file.
It contains multiple lines.
Python file handling is powerful.
Line 5: End of sample content."""

# Write the sample file
with open('sample.txt', 'w') as file:
    file.write(sample_content)

print("Created sample.txt for demonstration")
```

```python
# Basic file reading
print("=== BASIC FILE READING ===")

# Read entire file at once
with open('sample.txt', 'r') as file:
    content = file.read()
    print("Reading entire file:")
    print(content)
    print("-" * 40)

# Read line by line
with open('sample.txt', 'r') as file:
    print("Reading line by line:")
    for line_number, line in enumerate(file, 1):
        print(f"Line {line_number}: {line.rstrip()}")
    print("-" * 40)

# Read all lines into a list
with open('sample.txt', 'r') as file:
    lines = file.readlines()
    print(f"All lines as list: {lines}")
    print(f"Number of lines: {len(lines)}")
```

```python
# File writing operations
print("\n=== FILE WRITING OPERATIONS ===")

# Write mode ('w') - overwrites existing content
with open('output1.txt', 'w') as file:
    file.write("This is the first line.\n")
    file.write("This is the second line.\n")
    file.write("This overwrites any existing content.\n")

# Append mode ('a') - adds to existing content
with open('output1.txt', 'a') as file:
    file.write("This line was appended.\n")
    file.write("Append mode preserves existing content.\n")

# Read and display the result
with open('output1.txt', 'r') as file:
    print("Content of output1.txt:")
    print(file.read())
```

```python
# File modes summary
print("\n=== FILE MODES SUMMARY ===")
modes = [
    {'mode': 'r', 'description': 'Read (default) - opens file for reading', 'pointer': 'beginning'},
    {'mode': 'w', 'description': 'Write - creates/truncates file for writing', 'pointer': 'beginning'},
    {'mode': 'a', 'description': 'Append - opens file for appending', 'pointer': 'end'},
    {'mode': 'r+', 'description': 'Read+Write - opens file for both', 'pointer': 'beginning'},
    {'mode': 'w+', 'description': 'Read+Write - creates/truncates file', 'pointer': 'beginning'},
    {'mode': 'a+', 'description': 'Read+Append - opens file for both', 'pointer': 'end'}
]

print("File Mode Reference:")
print("Mode | Description | File Pointer Position")
print("-" * 50)
for mode_info in modes:
    print(f" {mode_info['mode']:>3} | {mode_info['description']:<35} | {mode_info['pointer']}")

# Demonstration of r+ mode
with open('sample.txt', 'r+') as file:
    # Read first line
    first_line = file.readline().strip()
    print(f"\nOriginal first line: '{first_line}'")
    
    # Go back to beginning and overwrite
    file.seek(0)
    file.write("MODIFIED: " + first_line + "\n")
    
    # Read the modified content
    file.seek(0)
    modified_first_line = file.readline().strip()
    print(f"Modified first line: '{modified_first_line}'")
```

## 🔒 Context Managers for Safe File Operations

Context managers ensure files are properly closed even if errors occur. The `with` statement is the recommended way to handle files.

```python
# Custom context manager example
class FileHandler:
    """Custom context manager for file operations."""
    
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None
    
    def __enter__(self):
        self.file = open(self.filename, self.mode)
        print(f"Opened file: {self.filename}")
        return self.file
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.file:
            self.file.close()
        print(f"Closed file: {self.filename}")
        return False  # Don't suppress exceptions

# Using custom context manager
print("=== CUSTOM CONTEXT MANAGER ===")
with FileHandler('context_demo.txt', 'w') as file:
    file.write("This file is managed by a custom context manager.\n")
    file.write("It will be automatically closed.\n")

# Verify the file was written and closed
with open('context_demo.txt', 'r') as file:
    print("Content written by custom context manager:")
    print(file.read())
```

```python
# Multiple file operations
print("\n=== MULTIPLE FILE OPERATIONS ===")

# Create multiple files with different content
files_data = {
    'file1.txt': 'Content of file 1',
    'file2.txt': 'Content of file 2',
    'file3.txt': 'Content of file 3'
}

# Write multiple files
for filename, content in files_data.items():
    with open(filename, 'w') as file:
        file.write(content)
    print(f"Created: {filename}")

# Read multiple files and combine content
combined_content = ""
for filename in files_data.keys():
    with open(filename, 'r') as file:
        combined_content += file.read() + "\n"

print(f"\nCombined content from all files:")
print(combined_content)
print(f"Total characters: {len(combined_content)}")
```

## 🗂️ Pathlib: Modern File System Operations

`pathlib` provides an object-oriented interface for filesystem operations, making them more intuitive and cross-platform.

```python
from pathlib import Path

print("=== PATHLIB OPERATIONS ===")

# Create Path objects
current_dir = Path.cwd()
data_dir = Path('data')
sample_file = Path('sample.txt')

print(f"Current directory: {current_dir}")
print(f"Data directory: {data_dir}")
print(f"Sample file: {sample_file}")
print(f"Sample file absolute: {sample_file.absolute()}")

# File properties
if sample_file.exists():
    print(f"\nFile exists: {sample_file.exists()}")
    print(f"Is file: {sample_file.is_file()}")
    print(f"Is directory: {sample_file.is_dir()}")
    print(f"File size: {sample_file.stat().st_size} bytes")
    print(f"Last modified: {sample_file.stat().st_mtime}")

# Directory operations
data_dir.mkdir(exist_ok=True)
subdir = data_dir / 'processed'
subdir.mkdir(exist_ok=True)

print(f"\nCreated directories: {data_dir}, {subdir}")

# List directory contents
print(f"\nCurrent directory contents:")
for item in current_dir.iterdir():
    if item.is_file():
        print(f"  📄 {item.name}")
    elif item.is_dir():
        print(f"  📁 {item.name}/")

# File operations with pathlib
source = Path('sample.txt')
destination = data_dir / 'sample_copy.txt'

if source.exists():
    import shutil
    shutil.copy2(source, destination)
    print(f"\nCopied {source} to {destination}")
```

```python
# Advanced pathlib operations
print("\n=== ADVANCED PATHLIB ===")

# Path components
complex_path = Path('/home/user/documents/projects/python/file.txt')
print(f"Full path: {complex_path}")
print(f"Name: {complex_path.name}")
print(f"Stem: {complex_path.stem}")
print(f"Suffix: {complex_path.suffix}")
print(f"Parent: {complex_path.parent}")
print(f"Parents: {list(complex_path.parents)}")

# Path manipulation
file_path = Path('data/raw/input.csv')
print(f"\nOriginal path: {file_path}")
print(f"With suffix changed: {file_path.with_suffix('.json')}")
print(f"With name changed: {file_path.with_name('output.csv')}")

# Pattern matching
print(f"\nFiles matching '*.txt' pattern:")
for txt_file in current_dir.glob('*.txt'):
    print(f"  {txt_file.name}")

print(f"\nAll files in data directory (recursive):")
for file_path in data_dir.rglob('*'):
    if file_path.is_file():
        print(f"  {file_path.relative_to(current_dir)}")
```

## 📄 Working with Different File Formats

Python provides excellent support for various file formats through its standard library.

```python
import json
import csv

# JSON operations
print("=== JSON FILE OPERATIONS ===")

# Sample data
user_data = {
    "name": "Alice Johnson",
    "age": 28,
    "city": "San Francisco",
    "skills": ["Python", "JavaScript", "SQL"],
    "projects": [
        {"name": "Web App", "status": "completed"},
        {"name": "API", "status": "in_progress"}
    ]
}

# Write JSON
with open('user_data.json', 'w') as json_file:
    json.dump(user_data, json_file, indent=2)

print("Created user_data.json")

# Read JSON
with open('user_data.json', 'r') as json_file:
    loaded_data = json.load(json_file)

print(f"Loaded user: {loaded_data['name']} from {loaded_data['city']}")
print(f"Skills: {loaded_data['skills']}")
```

```python
# CSV operations
print("\n=== CSV FILE OPERATIONS ===")

# Sample employee data
employees = [
    ["Name", "Department", "Salary", "Years"],
    ["Alice", "Engineering", "95000", "3"],
    ["Bob", "Marketing", "65000", "5"],
    ["Charlie", "Engineering", "85000", "2"],
    ["Diana", "HR", "55000", "7"]
]

# Write CSV
with open('employees.csv', 'w', newline='') as csv_file:
    writer = csv.writer(csv_file)
    for row in employees:
        writer.writerow(row)

print("Created employees.csv")

# Read CSV
with open('employees.csv', 'r') as csv_file:
    reader = csv.reader(csv_file)
    print("\nEmployee data:")
    for row in reader:
        print(f"  {', '.join(row)}")

# Read CSV with DictReader
with open('employees.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    engineering_employees = [
        row for row in reader
        if row['Department'] == 'Engineering'
    ]

print(f"\nEngineering employees: {len(engineering_employees)}")
for emp in engineering_employees:
    print(f"  {emp['Name']}: ${emp['Salary']}")
```

```python
# Binary file operations
print("\n=== BINARY FILE OPERATIONS ===")

# Write binary data
binary_data = bytes([0x41, 0x42, 0x43, 0x00, 0xFF, 0xFE])  # ABC + null + bytes

with open('binary_data.bin', 'wb') as binary_file:
    binary_file.write(binary_data)

print("Created binary_data.bin")

# Read binary data
with open('binary_data.bin', 'rb') as binary_file:
    data = binary_file.read()
    print(f"Binary data: {data}")
    print(f"As hex: {data.hex()}")
    print(f"As list: {list(data)}")

# Working with structured binary data
import struct

# Pack data into binary format
packed_data = struct.pack('i f d', 42, 3.14, 2.71828)  # int, float, double

with open('structured_data.bin', 'wb') as file:
    file.write(packed_data)

# Unpack data from binary format
with open('structured_data.bin', 'rb') as file:
    data = file.read()
    unpacked = struct.unpack('i f d', data)
    print(f"\nUnpacked data: {unpacked}")
    print(f"Integer: {unpacked[0]}, Float: {unpacked[1]:.2f}, Double: {unpacked[2]:.5f}")
```

## 🔍 File System Operations and Utilities

Advanced file system operations for robust applications.

```python
import os
import shutil
import tempfile

print("=== FILE SYSTEM OPERATIONS ===")

# Create project structure
project_dir = Path('my_project')
project_dir.mkdir(exist_ok=True)

# Create subdirectories
dirs_to_create = [
    project_dir / 'src',
    project_dir / 'tests',
    project_dir / 'docs',
    project_dir / 'data'
]

for dir_path in dirs_to_create:
    dir_path.mkdir(exist_ok=True)

# Create files
(project_dir / 'src' / 'main.py').write_text('# Main application file\nprint("Hello from main.py")')
(project_dir / 'src' / '__init__.py').write_text('# Package initialization')
(project_dir / 'README.md').write_text('# My Project\n\nA sample Python project.')
(project_dir / 'data' / 'sample.txt').write_text('Sample data file\nWith multiple lines.')

print(f"Created project structure at: {project_dir.absolute()}")

# Directory tree visualization
def print_tree(path, prefix=""):
    if path.is_file():
        print(f"{prefix}📄 {path.name}")
    elif path.is_dir():
        print(f"{prefix}📁 {path.name}/")
        for item in sorted(path.iterdir()):
            print_tree(item, prefix + "  ")

print("\nProject tree:")
print_tree(project_dir)
```

```python
# File statistics and operations
print("\n=== FILE STATISTICS ===")

# Calculate project statistics
total_files = sum(1 for _ in project_dir.rglob('*') if _.is_file())
total_dirs = sum(1 for _ in project_dir.rglob('*') if _.is_dir())
total_size = sum(f.stat().st_size for f in project_dir.rglob('*') if f.is_file())

print(f"Project statistics:")
print(f"  Directories: {total_dirs}")
print(f"  Files: {total_files}")
print(f"  Total size: {total_size} bytes")

# File type analysis
file_types = {}
for file_path in project_dir.rglob('*'):
    if file_path.is_file():
        ext = file_path.suffix or 'no_extension'
        file_types[ext] = file_types.get(ext, 0) + 1

print(f"\nFile types:")
for ext, count in sorted(file_types.items()):
    print(f"  {ext}: {count} files")

# Temporary files
print(f"\n=== TEMPORARY FILES ===")

with tempfile.NamedTemporaryFile(mode='w', suffix='.tmp', delete=False) as temp_file:
    temp_file.write("This is temporary data\n")
    temp_file.write("It will be automatically cleaned up\n")
    temp_filename = temp_file.name

print(f"Created temporary file: {temp_filename}")

# Read the temporary file
with open(temp_filename, 'r') as temp_file:
    print("Content of temporary file:")
    print(temp_file.read())

# Clean up
os.unlink(temp_filename)
print(f"Cleaned up temporary file: {temp_filename}")
```

## 🚨 Error Handling for File Operations

Robust file operations require comprehensive error handling.

```python
print("=== ERROR HANDLING FOR FILES ===")

def safe_file_operations():
    """Demonstrate safe file operations with error handling."""
    
    # Test files that may not exist
    test_files = [
        'existing_file.txt',
        'nonexistent.txt',
        'readonly_file.txt',
        'corrupt_file.json'
    ]
    
    # Create some test files
    Path('existing_file.txt').write_text('This file exists!')
    Path('corrupt_file.json').write_text('{invalid json content')
    
    for filename in test_files:
        print(f"\nProcessing {filename}:")
        
        try:
            # Try to read as text file
            with open(filename, 'r', encoding='utf-8') as file:
                content = file.read()
                print(f"  ✓ Successfully read {len(content)} characters")
                
                # Try to parse as JSON if it looks like JSON
                if filename.endswith('.json'):
                    try:
                        data = json.loads(content)
                        print(f"  ✓ Successfully parsed JSON: {type(data)}")
                    except json.JSONDecodeError as e:
                        print(f"  ⚠ JSON parsing failed: {e}")
                        
        except FileNotFoundError:
            print(f"  ✗ File not found: {filename}")
            
        except PermissionError:
            print(f"  ✗ Permission denied: {filename}")
            
        except UnicodeDecodeError as e:
            print(f"  ✗ Encoding error: {e}")
            
        except Exception as e:
            print(f"  ✗ Unexpected error: {type(e).__name__}: {e}")

safe_file_operations()

# Clean up test files
for filename in ['existing_file.txt', 'corrupt_file.json']:
    if Path(filename).exists():
        Path(filename).unlink()
```
