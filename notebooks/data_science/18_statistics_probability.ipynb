## 🎲 Probability Fundamentals

Probability is the foundation of statistical inference and data science.

```python
# Probability Fundamentals and Basic Concepts
print("=== PROBABILITY FUNDAMENTALS ===")

# Install required packages
try:
    import scipy.stats as stats
    import numpy as np
    import matplotlib.pyplot as plt
    import pandas as pd
    from statsmodels.stats.proportion import proportions_ztest
    from statsmodels.stats.weightstats import ztest
    import seaborn as sns
    STATS_AVAILABLE = True
except ImportError:
    # Dependency missing? Install once with: pip install -r requirements.txt
    import scipy.stats as stats
    import numpy as np
    import matplotlib.pyplot as plt
    import pandas as pd
    from statsmodels.stats.proportion import proportions_ztest
    from statsmodels.stats.weightstats import ztest
    import seaborn as sns
    STATS_AVAILABLE = True

if STATS_AVAILABLE:
    print("✓ Statistical packages successfully imported")
    
    # Set random seed for reproducibility
    np.random.seed(42)
    
    # Basic probability concepts
    print("\n1. BASIC PROBABILITY CONCEPTS:")
    
    # Sample space and events
    # Example: Rolling a fair six-sided die
    sample_space = [1, 2, 3, 4, 5, 6]
    print(f"Sample space (6-sided die): {sample_space}")
    print(f"Probability of rolling a 4: {1/6:.3f}")
    print(f"Probability of rolling an even number: {3/6:.3f}")
    print(f"Probability of rolling a number > 3: {3/6:.3f}")
    
    # Simulate die rolls
    n_rolls = 10000
    die_rolls = np.random.randint(1, 7, n_rolls)
    
    # Calculate empirical probabilities
    empirical_probs = {}
    for i in range(1, 7):
        empirical_probs[i] = np.sum(die_rolls == i) / n_rolls
    
    print(f"\nEmpirical probabilities from {n_rolls:,} rolls:")
    for face, prob in empirical_probs.items():
        print(f"  Face {face}: {prob:.4f} (theoretical: {1/6:.4f})")
    
    # Conditional probability
    print("\n2. CONDITIONAL PROBABILITY:")
    
    # Example: Drawing cards from a deck
    # P(King | Red) = P(King and Red) / P(Red)
    
    # Total cards: 52
    # Red cards: 26 (13 hearts + 13 diamonds)
    # Red kings: 2 (king of hearts, king of diamonds)
    
    p_red = 26/52  # Probability of red card
    p_king_and_red = 2/52  # Probability of red king
    p_king_given_red = p_king_and_red / p_red  # Conditional probability
    
    print(f"P(Red card): {p_red:.3f}")
    print(f"P(Red King): {p_king_and_red:.3f}")
    print(f"P(King|Red): {p_king_given_red:.3f}")
    
    # Bayes' Theorem
    print("\n3. BAYES' THEOREM:")
    
    # Example: Medical test
    # Disease prevalence: 1%
    # Test sensitivity: 99% (true positive rate)
    # Test specificity: 95% (true negative rate)
    
    p_disease = 0.01  # Prior probability
    p_positive_given_disease = 0.99  # Sensitivity
    p_negative_given_no_disease = 0.95  # Specificity
    p_positive_given_no_disease = 1 - p_negative_given_no_disease  # False positive rate
    
    # P(Disease|Positive) = P(Positive|Disease) * P(Disease) / P(Positive)
    p_positive = (p_positive_given_disease * p_disease + 
                  p_positive_given_no_disease * (1 - p_disease))
    
    p_disease_given_positive = (p_positive_given_disease * p_disease) / p_positive
    
    print(f"Disease prevalence: {p_disease:.1%}")
    print(f"Test sensitivity: {p_positive_given_disease:.1%}")
    print(f"Test specificity: {p_negative_given_no_disease:.1%}")
    print(f"P(Disease|Positive test): {p_disease_given_positive:.1%}")
    print("⚠️  Even with 99% sensitivity, only ~9% chance of having disease when test is positive!")
    
    # Random variables and distributions
    print("\n4. PROBABILITY DISTRIBUTIONS:")
    
    # Discrete uniform distribution (fair die)
    die_values = [1, 2, 3, 4, 5, 6]
    die_probs = [1/6] * 6
    
    plt.figure(figsize=(15, 10))
    
    # Plot discrete uniform
    plt.subplot(2, 3, 1)
    plt.bar(die_values, die_probs, alpha=0.7, color='blue', edgecolor='black')
    plt.title('Discrete Uniform Distribution\n(Fair 6-sided Die)')
    plt.xlabel('Face Value')
    plt.ylabel('Probability')
    plt.ylim(0, 0.2)
    
    # Normal distribution
    x_normal = np.linspace(-4, 4, 100)
    y_normal = stats.norm.pdf(x_normal, 0, 1)
    
    plt.subplot(2, 3, 2)
    plt.plot(x_normal, y_normal, 'r-', linewidth=2, label='μ=0, σ=1')
    plt.fill_between(x_normal, y_normal, alpha=0.3, color='red')
    plt.title('Normal Distribution')
    plt.xlabel('Value')
    plt.ylabel('Probability Density')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Binomial distribution
    n, p = 10, 0.3  # 10 trials, 30% success probability
    x_binom = np.arange(0, n+1)
    y_binom = stats.binom.pmf(x_binom, n, p)
    
    plt.subplot(2, 3, 3)
    plt.bar(x_binom, y_binom, alpha=0.7, color='green', edgecolor='black')
    plt.title(f'Binomial Distribution\n(n={n}, p={p})')
    plt.xlabel('Number of Successes')
    plt.ylabel('Probability')
    
    # Poisson distribution
    lambda_poisson = 3
    x_poisson = np.arange(0, 15)
    y_poisson = stats.poisson.pmf(x_poisson, lambda_poisson)
    
    plt.subplot(2, 3, 4)
    plt.bar(x_poisson, y_poisson, alpha=0.7, color='orange', edgecolor='black')
    plt.title(f'Poisson Distribution\n(λ={lambda_poisson})')
    plt.xlabel('Number of Events')
    plt.ylabel('Probability')
    
    # Exponential distribution
    x_exp = np.linspace(0, 5, 100)
    y_exp = stats.expon.pdf(x_exp, scale=1)
    
    plt.subplot(2, 3, 5)
    plt.plot(x_exp, y_exp, 'purple', linewidth=2, label='λ=1')
    plt.fill_between(x_exp, y_exp, alpha=0.3, color='purple')
    plt.title('Exponential Distribution')
    plt.xlabel('Time')
    plt.ylabel('Probability Density')
    plt.grid(True, alpha=0.3)
    
    # Chi-squared distribution
    x_chi2 = np.linspace(0, 20, 100)
    y_chi2 = stats.chi2.pdf(x_chi2, df=5)
    
    plt.subplot(2, 3, 6)
    plt.plot(x_chi2, y_chi2, 'brown', linewidth=2, label='df=5')
    plt.fill_between(x_chi2, y_chi2, alpha=0.3, color='brown')
    plt.title('Chi-squared Distribution')
    plt.xlabel('Value')
    plt.ylabel('Probability Density')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
else:
    print("Statistical packages not available")
```

## 📊 Descriptive Statistics

Descriptive statistics help us understand and summarize data.

```python
# Descriptive Statistics and Data Summarization
print("=== DESCRIPTIVE STATISTICS ===")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Create sample datasets
np.random.seed(42)

# Generate employee salary data
n_employees = 1000
ages = np.random.normal(35, 8, n_employees).clip(22, 65)
experience = np.random.normal(12, 6, n_employees).clip(0, 35)
base_salaries = 30000 + 2000 * experience + np.random.normal(0, 8000, n_employees)

# Add some outliers
outlier_indices = np.random.choice(n_employees, 20, replace=False)
base_salaries[outlier_indices[:10]] *= 3  # Executive salaries
base_salaries[outlier_indices[10:]] *= 0.3  # Entry-level

# Create DataFrame
df_salary = pd.DataFrame({
    'age': ages,
    'experience_years': experience,
    'salary': base_salaries
})

print("1. DATASET OVERVIEW:")
print(f"Dataset shape: {df_salary.shape}")
print("\nBasic statistics:")
print(df_salary.describe().round(2))

# Measures of central tendency
print("\n2. MEASURES OF CENTRAL TENDENCY:")
for col in ['age', 'experience_years', 'salary']:
    data = df_salary[col]
    mean_val = data.mean()
    median_val = data.median()
    mode_val = data.mode().iloc[0]  # Get first mode if multiple
    
    print(f"\n{col.upper()}:")
    print(f"  Mean: ${mean_val:,.0f}" if col == 'salary' else f"  Mean: {mean_val:.1f}")
    print(f"  Median: ${median_val:,.0f}" if col == 'salary' else f"  Median: {median_val:.1f}")
    print(f"  Mode: ${mode_val:,.0f}" if col == 'salary' else f"  Mode: {mode_val:.1f}")
    
    # Check for skewness
    skewness = data.skew()
    print(f"  Skewness: {skewness:.3f} ({'Right-skewed' if skewness > 0.5 else 'Left-skewed' if skewness < -0.5 else 'Approximately symmetric'})")

# Measures of dispersion
print("\n3. MEASURES OF DISPERSION:")
for col in ['age', 'experience_years', 'salary']:
    data = df_salary[col]
    variance = data.var()
    std_dev = data.std()
    iqr = data.quantile(0.75) - data.quantile(0.25)
    cv = std_dev / data.mean()  # Coefficient of variation
    
    print(f"\n{col.upper()}:")
    print(f"  Variance: {variance:,.0f}" if col == 'salary' else f"  Variance: {variance:.2f}")
    print(f"  Standard Deviation: {std_dev:,.0f}" if col == 'salary' else f"  Standard Deviation: {std_dev:.2f}")
    print(f"  IQR: {iqr:,.0f}" if col == 'salary' else f"  IQR: {iqr:.2f}")
    print(f"  Coefficient of Variation: {cv:.3f}")

# Visual exploration
plt.figure(figsize=(16, 12))

# Histograms
plt.subplot(3, 3, 1)
plt.hist(df_salary['age'], bins=30, alpha=0.7, color='blue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 2)
plt.hist(df_salary['experience_years'], bins=30, alpha=0.7, color='green', edgecolor='black')
plt.title('Experience Distribution')
plt.xlabel('Years of Experience')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 3)
plt.hist(df_salary['salary'], bins=50, alpha=0.7, color='red', edgecolor='black')
plt.title('Salary Distribution (with outliers)')
plt.xlabel('Salary ($)')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Box plots
plt.subplot(3, 3, 4)
sns.boxplot(y=df_salary['age'], color='blue')
plt.title('Age Box Plot')
plt.ylabel('Age')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 5)
sns.boxplot(y=df_salary['experience_years'], color='green')
plt.title('Experience Box Plot')
plt.ylabel('Years of Experience')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 6)
sns.boxplot(y=df_salary['salary'], color='red')
plt.title('Salary Box Plot (showing outliers)')
plt.ylabel('Salary ($)')
plt.grid(True, alpha=0.3)

# Q-Q plots (quantile-quantile)
plt.subplot(3, 3, 7)
stats.probplot(df_salary['age'], dist="norm", plot=plt)
plt.title('Age Q-Q Plot')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 8)
stats.probplot(df_salary['experience_years'], dist="norm", plot=plt)
plt.title('Experience Q-Q Plot')
plt.grid(True, alpha=0.3)

plt.subplot(3, 3, 9)
stats.probplot(df_salary['salary'], dist="norm", plot=plt)
plt.title('Salary Q-Q Plot\n(Shows deviation from normality)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Correlation analysis
print("\n4. CORRELATION ANALYSIS:")
correlation_matrix = df_salary.corr()
print("Correlation matrix:")
print(correlation_matrix.round(3))

# Visualize correlations
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

print("\n✓ Descriptive statistics analysis completed!")
print("✓ Data distributions, central tendency, and relationships explored")
```

## 🔬 Inferential Statistics: Hypothesis Testing

Hypothesis testing allows us to make statistical inferences about populations from samples.

```python
# Hypothesis Testing and Inferential Statistics
print("=== HYPOTHESIS TESTING ===")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
from statsmodels.stats.weightstats import ztest
from scipy.stats import ttest_ind, ttest_rel, chi2_contingency

# Create sample data for hypothesis testing
np.random.seed(42)

# Example 1: One-sample t-test
print("1. ONE-SAMPLE T-TEST:")
print("   H₀: Average employee age = 35 years")
print("   H₁: Average employee age ≠ 35 years")
print("   α = 0.05")

# Generate sample data
sample_ages = np.random.normal(36.5, 8, 100)  # True mean is 36.5
hypothesized_mean = 35

# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(sample_ages, hypothesized_mean)

print(f"\nSample mean: {sample_ages.mean():.2f} years")
print(f"Hypothesized mean: {hypothesized_mean} years")
print(f"T-statistic: {t_statistic:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Reject H₀: {p_value < 0.05} (p < 0.05)")

if p_value < 0.05:
    print("   📊 CONCLUSION: There is statistically significant evidence that the average age differs from 35 years")
else:
    print("   📊 CONCLUSION: No statistically significant evidence that the average age differs from 35 years")

# Example 2: Two-sample t-test
print("\n\n2. TWO-SAMPLE T-TEST:")
print("   H₀: Average salary is the same for male and female employees")
print("   H₁: Average salary differs between male and female employees")

# Generate salary data for two groups
male_salaries = np.random.normal(55000, 12000, 150)
female_salaries = np.random.normal(52000, 11000, 130)  # Slightly lower average

print(f"\nMale salaries - Mean: ${male_salaries.mean():.0f}, Std: ${male_salaries.std():.0f}, N: {len(male_salaries)}")
print(f"Female salaries - Mean: ${female_salaries.mean():.0f}, Std: ${female_salaries.std():.0f}, N: {len(female_salaries)}")

# Perform two-sample t-test
t_stat, p_value = ttest_ind(male_salaries, female_salaries)

print(f"\nT-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Reject H₀: {p_value < 0.05} (p < 0.05)")

if p_value < 0.05:
    print("   📊 CONCLUSION: There is statistically significant evidence of salary difference between genders")
else:
    print("   📊 CONCLUSION: No statistically significant evidence of salary difference between genders")

# Example 3: Chi-square test for independence
print("\n\n3. CHI-SQUARE TEST FOR INDEPENDENCE:")
print("   H₀: Job satisfaction is independent of department")
print("   H₁: Job satisfaction depends on department")

# Create contingency table
# Rows: Departments, Columns: Satisfaction levels
contingency_table = np.array([
    [45, 25, 10],  # Engineering: Satisfied, Neutral, Dissatisfied
    [35, 30, 15],  # Marketing: Satisfied, Neutral, Dissatisfied
    [40, 20, 20],  # Sales: Satisfied, Neutral, Dissatisfied
    [50, 15, 5]    # HR: Satisfied, Neutral, Dissatisfied
])

departments = ['Engineering', 'Marketing', 'Sales', 'HR']
satisfaction_levels = ['Satisfied', 'Neutral', 'Dissatisfied']

print("\nContingency table:")
df_contingency = pd.DataFrame(contingency_table, 
                             index=departments, 
                             columns=satisfaction_levels)
print(df_contingency)

# Perform chi-square test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-square statistic: {chi2_stat:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Reject H₀: {p_value < 0.05} (p < 0.05)")

if p_value < 0.05:
    print("   📊 CONCLUSION: Job satisfaction is significantly associated with department")
else:
    print("   📊 CONCLUSION: No significant association between job satisfaction and department")

# Example 4: Confidence intervals
print("\n\n4. CONFIDENCE INTERVALS:")

# Calculate 95% confidence interval for mean salary
all_salaries = np.concatenate([male_salaries, female_salaries])
sample_mean = all_salaries.mean()
sample_std = all_salaries.std()
sample_size = len(all_salaries)

# Standard error
standard_error = sample_std / np.sqrt(sample_size)

# 95% confidence interval (t-distribution for sample)
confidence_level = 0.95
t_critical = stats.t.ppf((1 + confidence_level) / 2, df=sample_size-1)
margin_of_error = t_critical * standard_error

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"Sample mean salary: ${sample_mean:.0f}")
print(f"Sample size: {sample_size}")
print(f"95% Confidence Interval: ${ci_lower:.0f} to ${ci_upper:.0f}")
print(f"Margin of error: ${margin_of_error:.0f}")
print(f"We are 95% confident that the true population mean salary lies within this interval")

# Visualize confidence intervals
plt.figure(figsize=(14, 10))

# Plot 1: Hypothesis testing visualization
plt.subplot(2, 3, 1)
# Create normal distribution under null hypothesis
x_hyp = np.linspace(32, 38, 100)
y_hyp = stats.norm.pdf(x_hyp, hypothesized_mean, sample_ages.std()/np.sqrt(len(sample_ages)))
plt.plot(x_hyp, y_hyp, 'b-', linewidth=2, label='Null Distribution')

# Plot sample mean
sample_mean_age = sample_ages.mean()
plt.axvline(sample_mean_age, color='red', linestyle='--', linewidth=2, 
           label=f'Sample Mean: {sample_mean_age:.2f}')
plt.axvline(hypothesized_mean, color='green', linestyle='-', linewidth=2,
           label=f'H₀ Mean: {hypothesized_mean}')

# Shade critical regions (two-tailed test)
critical_value = stats.t.ppf(0.975, len(sample_ages)-1) * (sample_ages.std()/np.sqrt(len(sample_ages)))
plt.axvline(hypothesized_mean - critical_value, color='gray', linestyle=':', alpha=0.7)
plt.axvline(hypothesized_mean + critical_value, color='gray', linestyle=':', alpha=0.7)
plt.fill_between(x_hyp, y_hyp, where=(x_hyp < hypothesized_mean - critical_value) | (x_hyp > hypothesized_mean + critical_value), 
                alpha=0.3, color='red', label='Rejection Region')

plt.title('One-Sample T-Test Visualization')
plt.xlabel('Age')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 2: Box plot comparison for two-sample test
plt.subplot(2, 3, 2)
salary_data = pd.DataFrame({
    'Salary': np.concatenate([male_salaries, female_salaries]),
    'Gender': ['Male'] * len(male_salaries) + ['Female'] * len(female_salaries)
})

sns.boxplot(x='Gender', y='Salary', data=salary_data)
plt.title('Salary Comparison by Gender')
plt.ylabel('Salary ($)')
plt.grid(True, alpha=0.3)

# Plot 3: Contingency table heatmap
plt.subplot(2, 3, 3)
sns.heatmap(df_contingency, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Department vs Satisfaction\n(Chi-square Test)')
plt.xlabel('Satisfaction Level')
plt.ylabel('Department')

# Plot 4: Confidence interval visualization
plt.subplot(2, 3, 4)
# Sample from different populations to show CI variation
sample_sizes = [30, 100, 500]
colors = ['red', 'blue', 'green']

for i, n in enumerate(sample_sizes):
    sample = np.random.normal(sample_mean, sample_std, n)
    sample_mean_est = sample.mean()
    sample_std_est = sample.std()
    
    # Confidence interval
    se = sample_std_est / np.sqrt(n)
    t_crit = stats.t.ppf(0.975, n-1)
    ci_lower = sample_mean_est - t_crit * se
    ci_upper = sample_mean_est + t_crit * se
    
    plt.errorbar(i, sample_mean_est, yerr=t_crit*se, fmt='o', 
                color=colors[i], capsize=5, label=f'n={n}')
    
    # Mark true population mean
    plt.axhline(sample_mean, color='black', linestyle='--', alpha=0.7, label='True Mean' if i == 0 else "")

plt.xticks(range(len(sample_sizes)), [f'n={n}' for n in sample_sizes])
plt.ylabel('Estimated Mean Salary ($)')
plt.title('Confidence Intervals by Sample Size')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 5: Power analysis concept
plt.subplot(2, 3, 5)
# Show effect size and sample size relationship
effect_sizes = np.linspace(0.1, 1.0, 50)
sample_sizes_power = np.linspace(20, 200, 50)

# Simplified power calculation concept
power_mesh = np.zeros((len(sample_sizes_power), len(effect_sizes)))
for i, n in enumerate(sample_sizes_power):
    for j, effect in enumerate(effect_sizes):
        # Simplified power approximation
        se = 1 / np.sqrt(n)  # Assuming unit variance
        z_score = effect / se
        power_mesh[i, j] = 1 - stats.norm.cdf(1.96 - z_score)  # Approximation

plt.contourf(effect_sizes, sample_sizes_power, power_mesh, levels=np.linspace(0, 1, 11), cmap='RdYlGn')
plt.colorbar(label='Statistical Power')
plt.xlabel('Effect Size')
plt.ylabel('Sample Size')
plt.title('Statistical Power Analysis')
plt.grid(True, alpha=0.3)

# Plot 6: Common statistical pitfalls
plt.subplot(2, 3, 6)
plt.text(0.1, 0.9, '⚠️ Common Statistical Pitfalls:', fontsize=12, fontweight='bold')
plt.text(0.1, 0.8, '• Correlation ≠ Causation', fontsize=10)
plt.text(0.1, 0.7, '• Statistical vs Practical Significance', fontsize=10)
plt.text(0.1, 0.6, '• p-hacking (multiple testing)', fontsize=10)
plt.text(0.1, 0.5, '• Simpson\'s Paradox', fontsize=10)
plt.text(0.1, 0.4, '• Confounding Variables', fontsize=10)
plt.text(0.1, 0.3, '• Sample Selection Bias', fontsize=10)
plt.text(0.1, 0.2, '• Data Snooping', fontsize=10)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.axis('off')
plt.title('Statistical Pitfalls to Avoid')

plt.tight_layout()
plt.show()

print("\n✓ Hypothesis testing analysis completed!")
print("✓ One-sample, two-sample, and chi-square tests performed")
print("✓ Confidence intervals calculated and visualized")
print("✓ Common statistical pitfalls highlighted")
```

## 🎯 A/B Testing and Experimental Design

A/B testing is a crucial skill for data-driven decision making.

```python
# A/B Testing and Experimental Design
print("=== A/B TESTING AND EXPERIMENTAL DESIGN ===")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
from statsmodels.stats.proportion import proportions_ztest
from statsmodels.stats.weightstats import ztest

# Simulate A/B test data
np.random.seed(42)

# Example: Website conversion rate optimization
print("1. A/B TEST EXAMPLE: WEBSITE CONVERSION RATE")
print("   Testing: New website design vs Original design")
print("   Metric: Conversion rate (purchase completion)")
print("   H₀: Conversion rates are equal")
print("   H₁: New design has different conversion rate")

# Generate synthetic data
n_control = 5000  # Original design visitors
n_treatment = 5000  # New design visitors

# Control group: 3.5% conversion rate
# Treatment group: 4.2% conversion rate (16.7% improvement)
control_conversions = np.random.binomial(n_control, 0.035)
treatment_conversions = np.random.binomial(n_treatment, 0.042)

control_rate = control_conversions / n_control
treatment_rate = treatment_conversions / n_treatment

print(f"\nControl group (Original): {control_conversions}/{n_control} = {control_rate:.3%} conversion rate")
print(f"Treatment group (New): {treatment_conversions}/{n_treatment} = {treatment_rate:.3%} conversion rate")
print(f"Absolute improvement: {(treatment_rate - control_rate):.3%}")
print(f"Relative improvement: {((treatment_rate - control_rate) / control_rate):.1%}")

# Perform statistical test for proportions
successes = np.array([treatment_conversions, control_conversions])
n_obs = np.array([n_treatment, n_control])

z_stat, p_value = proportions_ztest(successes, n_obs)

print(f"\nStatistical Test Results:")
print(f"Z-statistic: {z_stat:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Statistically significant: {p_value < 0.05} (α = 0.05)")

if p_value < 0.05:
    print("   📊 RESULT: The new design shows statistically significant improvement!")
else:
    print("   📊 RESULT: No statistically significant difference detected.")

# Calculate confidence intervals
def proportion_confidence_interval(successes, n, confidence_level=0.95):
    """Calculate confidence interval for proportion"""
    p_hat = successes / n
    se = np.sqrt(p_hat * (1 - p_hat) / n)
    z = stats.norm.ppf((1 + confidence_level) / 2)
    margin = z * se
    return p_hat - margin, p_hat + margin

control_ci_lower, control_ci_upper = proportion_confidence_interval(control_conversions, n_control)
treatment_ci_lower, treatment_ci_upper = proportion_confidence_interval(treatment_conversions, n_treatment)

print(f"\nConfidence Intervals (95%):")
print(f"Control: {control_rate:.3%} ± {(control_ci_upper - control_rate):.3%} → [{control_ci_lower:.3%}, {control_ci_upper:.3%}]")
print(f"Treatment: {treatment_rate:.3%} ± {(treatment_ci_upper - treatment_rate):.3%} → [{treatment_ci_lower:.3%}, {treatment_ci_upper:.3%}]")

# Check if confidence intervals overlap
intervals_overlap = control_ci_upper >= treatment_ci_lower and treatment_ci_upper >= control_ci_lower
print(f"Confidence intervals overlap: {intervals_overlap}")

# Power analysis
print("\n2. POWER ANALYSIS:")
print("   What sample size do we need for reliable A/B tests?")

# Calculate required sample size for given effect size
def calculate_sample_size(p1, p2, alpha=0.05, power=0.8):
    """Calculate sample size for A/B test"""
    # Simplified calculation
    p_avg = (p1 + p2) / 2
    effect_size = abs(p1 - p2)
    
    # Z-scores
    z_alpha = stats.norm.ppf(1 - alpha/2)
    z_power = stats.norm.ppf(power)
    
    # Sample size formula
    n = (z_alpha * np.sqrt(2 * p_avg * (1 - p_avg)) + z_power * np.sqrt(p1 * (1 - p1) + p2 * (1 - p2)))**2 / effect_size**2
    return int(np.ceil(n))

# Different scenarios
scenarios = [
    (0.035, 0.042, "Small improvement"),
    (0.035, 0.05, "Medium improvement"),
    (0.035, 0.07, "Large improvement")
]

print("Required sample size per group (80% power, 95% confidence):")
for p1, p2, desc in scenarios:
    n_required = calculate_sample_size(p1, p2)
    print(f"   {desc} ({(p2/p1-1):.1%} lift): {n_required:,} users per group")

# Example 2: Revenue per user A/B test
print("\n\n3. REVENUE PER USER A/B TEST:")
print("   Testing: New pricing strategy vs Original pricing")
print("   Metric: Average revenue per user (ARPU)")

# Generate revenue data
n_users = 2000
control_revenue = np.random.lognormal(mean=4.5, sigma=0.8, size=n_users)  # ~$90 average
treatment_revenue = np.random.lognormal(mean=4.7, sigma=0.8, size=n_users)  # ~$110 average (22% increase)

control_arpu = control_revenue.mean()
treatment_arpu = treatment_revenue.mean()

print(f"\nControl ARPU: ${control_arpu:.2f}")
print(f"Treatment ARPU: ${treatment_arpu:.2f}")
print(f"Revenue increase: ${(treatment_arpu - control_arpu):.2f} ({((treatment_arpu/control_arpu)-1):.1%})")

# Test for difference in means
t_stat, p_value = stats.ttest_ind(treatment_revenue, control_revenue)

print(f"\nT-test results:")
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"Statistically significant: {p_value < 0.05}")

if p_value < 0.05:
    print("   📊 RESULT: New pricing strategy significantly increases revenue!")
else:
    print("   📊 RESULT: No significant difference in revenue detected.")

# Visualize A/B test results
plt.figure(figsize=(16, 12))

# Plot 1: Conversion rates with error bars
plt.subplot(3, 3, 1)
groups = ['Control', 'Treatment']
rates = [control_rate, treatment_rate]
errors = [(control_ci_upper - control_rate), (treatment_ci_upper - treatment_rate)]

bars = plt.bar(groups, rates, yerr=errors, capsize=5, color=['skyblue', 'orange'], alpha=0.7)
plt.ylabel('Conversion Rate')
plt.title('A/B Test: Conversion Rates\n(with 95% CI)')
plt.ylim(0, max(rates) + max(errors) + 0.005)

# Add value labels
for bar, rate in zip(bars, rates):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.001, 
             f'{rate:.3%}', ha='center', va='bottom', fontweight='bold')

plt.grid(True, alpha=0.3)

# Plot 2: Revenue distribution comparison
plt.subplot(3, 3, 2)
plt.hist(control_revenue, bins=50, alpha=0.7, label='Control', color='skyblue', density=True)
plt.hist(treatment_revenue, bins=50, alpha=0.7, label='Treatment', color='orange', density=True)
plt.xlabel('Revenue per User ($)')
plt.ylabel('Density')
plt.title('Revenue Distribution Comparison')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 3: Cumulative conversion rates
plt.subplot(3, 3, 3)
# Simulate cumulative results over time
days = np.arange(1, 31)
daily_control = np.random.binomial(n_control//30, control_rate, 30)
daily_treatment = np.random.binomial(n_treatment//30, treatment_rate, 30)

cum_control = np.cumsum(daily_control) / np.cumsum(np.full(30, n_control//30))
cum_treatment = np.cumsum(daily_treatment) / np.cumsum(np.full(30, n_treatment//30))

plt.plot(days, cum_control, 'b-', linewidth=2, label='Control')
plt.plot(days, cum_treatment, 'r-', linewidth=2, label='Treatment')
plt.xlabel('Day')
plt.ylabel('Cumulative Conversion Rate')
plt.title('A/B Test Progress Over Time')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 4: Statistical power curves
plt.subplot(3, 3, 4)
sample_sizes = np.logspace(2, 5, 50)  # 100 to 100,000
effect_sizes = [0.01, 0.02, 0.05]  # Small, medium, large effects
colors = ['blue', 'green', 'red']

for effect, color, label in zip(effect_sizes, colors, ['Small (1%)', 'Medium (2%)', 'Large (5%)']):
    # Simplified power calculation
    p_avg = 0.035
    se = np.sqrt(2 * p_avg * (1 - p_avg) / sample_sizes)
    z_scores = effect / se
    power = 1 - stats.norm.cdf(1.96 - z_scores)
    power = np.clip(power, 0, 1)  # Ensure between 0 and 1
    
    plt.semilogx(sample_sizes, power, color=color, linewidth=2, label=label)

plt.axhline(0.8, color='black', linestyle='--', alpha=0.7, label='Target Power (80%)')
plt.xlabel('Sample Size (per group)')
plt.ylabel('Statistical Power')
plt.title('Statistical Power vs Sample Size')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 5: P-value distribution under null
plt.subplot(3, 3, 5)
# Simulate p-values under null hypothesis
np.random.seed(42)
n_simulations = 10000
p_values_null = np.random.uniform(0, 1, n_simulations)

plt.hist(p_values_null, bins=50, alpha=0.7, color='purple', density=True, label='Null Distribution')
plt.axvline(0.05, color='red', linestyle='--', linewidth=2, label='α = 0.05')
plt.xlabel('P-value')
plt.ylabel('Density')
plt.title('P-value Distribution\n(Under Null Hypothesis)')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 6: Effect size interpretation
plt.subplot(3, 3, 6)
effect_sizes = np.linspace(0, 1, 100)
p_values_effect = [stats.ttest_ind(np.random.normal(0, 1, 100), 
                                  np.random.normal(effect, 1, 100))[1] 
                  for effect in effect_sizes]

# Simplified relationship
plt.plot(effect_sizes, [min(1, 1 - effect*2) for effect in effect_sizes], 
         'b-', linewidth=2, label='Simplified Relationship')
plt.xlabel('Effect Size (Cohen\'s d)')
plt.ylabel('Typical P-value')
plt.title('Effect Size vs Statistical Significance')
plt.grid(True, alpha=0.3)

# Plot 7: Common A/B testing mistakes
plt.subplot(3, 3, 7)
mistakes = [
    'Peeking at results early',
    'Not accounting for multiple testing',
    'Ignoring practical significance',
    'Testing too many variants',
    'Stopping tests too early',
    'Ignoring seasonality',
    'Not ensuring random assignment'
]

y_pos = np.arange(len(mistakes))
plt.barh(y_pos, [1]*len(mistakes), color='salmon', alpha=0.7)
plt.yticks(y_pos, mistakes)
plt.title('Common A/B Testing Mistakes')
plt.xlim(0, 1.2)

# Hide x-axis
plt.gca().axes.get_xaxis().set_visible(False)

# Plot 8: Confidence interval visualization
plt.subplot(3, 3, 8)
# Simulate multiple A/B test results
np.random.seed(42)
n_experiments = 20
true_effects = np.random.normal(0, 0.02, n_experiments)  # Some positive, some negative

experiment_results = []
for true_effect in true_effects:
    # Simulate experiment with measurement error
    measured_effect = true_effect + np.random.normal(0, 0.01)
    se = 0.015  # Standard error
    ci_lower = measured_effect - 1.96 * se
    ci_upper = measured_effect + 1.96 * se
    experiment_results.append((measured_effect, ci_lower, ci_upper, true_effect))

# Sort by measured effect
experiment_results.sort(key=lambda x: x[0])

for i, (measured, lower, upper, true) in enumerate(experiment_results):
    color = 'green' if lower <= 0 <= upper else 'red'
    plt.errorbar(measured, i, xerr=[[measured-lower], [upper-measured]], 
                fmt='o', color=color, capsize=3, alpha=0.7)
    plt.plot(true, i, 'kx', markersize=4, alpha=0.5)  # True effect

plt.axvline(0, color='black', linestyle='--', alpha=0.7, label='No Effect')
plt.xlabel('Effect Size')
plt.ylabel('Experiment')
plt.title('Multiple Experiment Results\n(Green: Statistically Significant)')
plt.grid(True, alpha=0.3)

# Plot 9: Sample size calculator visualization
plt.subplot(3, 3, 9)
baseline_rates = np.linspace(0.01, 0.1, 20)
improvement_rates = np.linspace(0.05, 0.5, 20)

# Calculate sample sizes for a grid
sample_size_grid = np.zeros((len(baseline_rates), len(improvement_rates)))
for i, baseline in enumerate(baseline_rates):
    for j, improvement in enumerate(improvement_rates):
        try:
            sample_size_grid[i, j] = calculate_sample_size(baseline, baseline * (1 + improvement))
        except:
            sample_size_grid[i, j] = 100000  # Large number for impossible calculations

# Plot as heatmap
plt.imshow(np.log10(sample_size_grid), cmap='YlOrRd_r', aspect='auto', origin='lower')
plt.colorbar(label='log₁₀(Sample Size)')
plt.xlabel('Relative Improvement')
plt.ylabel('Baseline Rate')
plt.title('Sample Size Requirements')

# Set tick labels
plt.xticks([0, len(improvement_rates)-1], ['5%', '50%'])
plt.yticks([0, len(baseline_rates)-1], ['1%', '10%'])

plt.tight_layout()
plt.show()

print("\n✓ A/B testing analysis completed!")
print("✓ Conversion rate and revenue A/B tests performed")
print("✓ Statistical power analysis demonstrated")
print("✓ Common A/B testing mistakes highlighted")
print("✓ Sample size calculations provided")
```

## 💡 Pro Tips

1. **Hypothesis Testing**: Always state your null and alternative hypotheses clearly
2. **P-values**: A p-value is the probability of observing your data (or more extreme) assuming H₀ is true
3. **Confidence Intervals**: These quantify the uncertainty in your estimates
4. **Power Analysis**: Determine sample size needs before running experiments
5. **Multiple Testing**: Correct for multiple comparisons to avoid false positives
6. **Effect Size**: Focus on practical significance, not just statistical significance
7. **Assumptions**: Always check if your data meets the assumptions of your statistical tests
8. **Visualization**: Plot your data distributions, not just summary statistics
9. **Context**: Consider the business/practical implications of your statistical findings
10. **Reproducibility**: Use set random seeds and document your analysis pipeline
