## ⚡ Understanding Asynchronous Programming

Asynchronous programming allows your program to handle multiple operations concurrently without blocking, improving performance and responsiveness.

````python
# Synchronous vs Asynchronous execution
print("=== SYNCHRONOUS VS ASYNCHRONOUS EXECUTION ===")

import time
import asyncio

# Synchronous function - blocks execution
def sync_task(name, duration):
    print(f"Starting {name}...")
    time.sleep(duration)  # This blocks!
    print(f"Finished {name} after {duration}s")
    return f"Result from {name}"

# Asynchronous function - doesn't block
async def async_task(name, duration):
    print(f"Starting {name}...")
    await asyncio.sleep(duration)  # This doesn't block!
    print(f"Finished {name} after {duration}s")
    return f"Result from {name}"

# Demonstrate synchronous execution
print("Synchronous execution:")
start_time = time.time()
result1 = sync_task("Task A", 1)
result2 = sync_task("Task B", 1)
result3 = sync_task("Task C", 1)
sync_time = time.time() - start_time
print(f"Synchronous total time: {sync_time:.2f}s\n")

# Demonstrate asynchronous execution
async def run_async_demo():
    print("Asynchronous execution:")
    start_time = time.time()
    
    # Run tasks concurrently
    results = await asyncio.gather(
        async_task("Task A", 1),
        async_task("Task B", 1),
        async_task("Task C", 1)
    )
    
    async_time = time.time() - start_time
    print(f"Asynchronous total time: {async_time:.2f}s")
    print(f"Time saved: {sync_time - async_time:.2f}s")
    return results

# Run the async demo
async_results = asyncio.run(run_async_demo())
print(f"Results: {async_results}")
````

````python
# Basic async/await syntax
print("\n=== BASIC ASYNC/AWAIT SYNTAX ===")

# Defining async functions
async def greet(name):
    """An asynchronous greeting function."""
    await asyncio.sleep(0.1)  # Simulate some async work
    return f"Hello, {name}!"

async def farewell(name):
    """An asynchronous farewell function."""
    await asyncio.sleep(0.1)  # Simulate some async work
    return f"Goodbye, {name}!"

# Using async functions
async def main():
    print("Starting async operations...")
    
    # Sequential execution (still async but not concurrent)
    greeting1 = await greet("Alice")
    farewell1 = await farewell("Alice")
    
    print(f"Sequential: {greeting1}, {farewell1}")
    
    # Concurrent execution
    greeting2, farewell2 = await asyncio.gather(
        greet("Bob"),
        farewell("Charlie")
    )
    
    print(f"Concurrent: {greeting2}, {farewell2}")
    
    # Multiple concurrent operations
    names = ["David", "Eve", "Frank"]
    greetings = await asyncio.gather(*[greet(name) for name in names])
    farewells = await asyncio.gather(*[farewell(name) for name in names])
    
    print(f"Multiple greetings: {greetings}")
    print(f"Multiple farewells: {farewells}")

# Run the async main function
asyncio.run(main())
````

````python
# Understanding coroutines and awaitables
print("\n=== COROUTINES AND AWAITABLES ===")

# Coroutine function
async def simple_coroutine():
    print("Coroutine started")
    await asyncio.sleep(0.5)
    print("Coroutine finished")
    return "Coroutine result"

# Task creation
async def task_demo():
    print("Creating tasks...")
    
    # Create tasks from coroutines
    task1 = asyncio.create_task(simple_coroutine())
    task2 = asyncio.create_task(simple_coroutine())
    task3 = asyncio.create_task(simple_coroutine())
    
    print(f"Tasks created: {len([task1, task2, task3])} tasks")
    print(f"Task states: {[task.done() for task in [task1, task2, task3]]}")
    
    # Wait for tasks to complete
    results = await asyncio.gather(task1, task2, task3)
    
    print(f"Task states after completion: {[task.done() for task in [task1, task2, task3]]}")
    print(f"Results: {results}")

# Inspecting awaitables
async def inspect_awaitables():
    coroutine = simple_coroutine()
    task = asyncio.create_task(simple_coroutine())
    
    print(f"Coroutine type: {type(coroutine)}")
    print(f"Task type: {type(task)}")
    print(f"Coroutine done: {coroutine.done() if hasattr(coroutine, 'done') else 'N/A'}")
    print(f"Task done: {task.done()}")
    
    # Execute the task
    result = await task
    print(f"Task result: {result}")
    print(f"Task done after execution: {task.done()}")

# Run demonstrations
asyncio.run(task_demo())
print()
asyncio.run(inspect_awaitables())
````

## 🔄 Event Loops and Task Management

The event loop is the core of async programming in Python, managing the execution of coroutines and tasks.

````python
# Event loop basics
print("=== EVENT LOOP BASICS ===")

async def event_loop_demo():
    print("Inside the event loop")
    
    # Get the current event loop
    loop = asyncio.get_event_loop()
    print(f"Event loop: {loop}")
    print(f"Event loop is running: {loop.is_running()}")
    
    # Schedule tasks
    async def delayed_print(message, delay):
        await asyncio.sleep(delay)
        print(f"Delayed message: {message}")
    
    # Create and schedule tasks
    tasks = [
        asyncio.create_task(delayed_print(f"Task {i}", i * 0.1))
        for i in range(1, 4)
    ]
    
    print(f"Created {len(tasks)} tasks")
    
    # Wait for all tasks to complete
    await asyncio.gather(*tasks)
    
    print("All tasks completed")

# Run event loop demo
asyncio.run(event_loop_demo())
````

````python
# Advanced task management
print("\n=== ADVANCED TASK MANAGEMENT ===")

async def task_management_demo():
    async def worker(name, duration):
        print(f"{name} starting work")
        await asyncio.sleep(duration)
        print(f"{name} finished work")
        return f"{name} result"
    
    print("=== Task Creation and Execution ===")
    # Create tasks
    task1 = asyncio.create_task(worker("Worker A", 1.0))
    task2 = asyncio.create_task(worker("Worker B", 0.5))
    task3 = asyncio.create_task(worker("Worker C", 1.5))
    
    print(f"Tasks created: {[task.get_name() for task in [task1, task2, task3]]}")
    
    # Wait for first completed task
    print("\n=== Waiting for First Completion ===")
    done, pending = await asyncio.wait([task1, task2, task3], return_when=asyncio.FIRST_COMPLETED)
    print(f"Completed tasks: {len(done)}, Pending tasks: {len(pending)}")
    
    # Cancel remaining tasks
    for task in pending:
        task.cancel()
    
    # Wait for cancelled tasks to actually cancel
    await asyncio.gather(*pending, return_exceptions=True)
    
    print("\n=== Task Results ===")
    for task in done:
        try:
            result = task.result()
            print(f"Task result: {result}")
        except asyncio.CancelledError:
            print("Task was cancelled")
    
    for task in pending:
        print(f"Pending task cancelled: {task.cancelled()}")

# Run task management demo
asyncio.run(task_management_demo())
````

````python
# Async context managers and iterators
print("\n=== ASYNC CONTEXT MANAGERS AND ITERATORS ===")

# Async context manager
class AsyncTimer:
    """An async context manager that measures execution time."""
    
    def __init__(self, name):
        self.name = name
        self.start_time = None
    
    async def __aenter__(self):
        self.start_time = asyncio.get_event_loop().time()
        print(f"Starting {self.name}...")
        return self
    
    async def __aexit__(self, exc_type, exc_val, exc_tb):
        end_time = asyncio.get_event_loop().time()
        duration = end_time - self.start_time
        print(f"Finished {self.name} in {duration:.3f}s")
        return False

# Async iterator
class AsyncCounter:
    """An async iterator that yields numbers with delays."""
    
    def __init__(self, start, end, delay=0.1):
        self.start = start
        self.end = end
        self.delay = delay
        self.current = start
    
    def __aiter__(self):
        return self
    
    async def __anext__(self):
        if self.current >= self.end:
            raise StopAsyncIteration
        
        value = self.current
        self.current += 1
        
        await asyncio.sleep(self.delay)
        return value

async def async_context_demo():
    # Use async context manager
    async with AsyncTimer("async operations"):
        # Simulate some async work
        await asyncio.sleep(0.2)
        
        # Use async iterator
        print("Counting asynchronously:")
        async for number in AsyncCounter(1, 6):
            print(f"Got number: {number}")
        
        await asyncio.sleep(0.1)

# Run async context demo
asyncio.run(async_context_demo())
````

## 🌐 Real-World Async Applications

Async programming excels at I/O-bound operations like network requests, file operations, and API calls.

````python
# Async HTTP requests simulation
print("=== ASYNC HTTP REQUESTS SIMULATION ===")

import random

# Simulate async HTTP client
class AsyncHTTPClient:
    """Simulated async HTTP client."""
    
    async def get(self, url):
        """Simulate an async HTTP GET request."""
        print(f"Fetching {url}...")
        
        # Simulate network latency (random 0.1-1.0 seconds)
        latency = random.uniform(0.1, 1.0)
        await asyncio.sleep(latency)
        
        # Simulate response
        response_data = {
            "url": url,
            "status": 200,
            "data": f"Response from {url}",
            "latency": latency
        }
        
        print(f"Completed {url} in {latency:.2f}s")
        return response_data

async def fetch_multiple_urls():
    """Fetch multiple URLs concurrently."""
    client = AsyncHTTPClient()
    urls = [
        "https://api.example.com/users",
        "https://api.example.com/posts",
        "https://api.example.com/comments",
        "https://api.example.com/products",
        "https://api.example.com/categories"
    ]
    
    print(f"Fetching {len(urls)} URLs concurrently...")
    start_time = asyncio.get_event_loop().time()
    
    # Fetch all URLs concurrently
    responses = await asyncio.gather(
        *[client.get(url) for url in urls],
        return_exceptions=True
    )
    
    end_time = asyncio.get_event_loop().time()
    total_time = end_time - start_time
    
    print(f"\nAll requests completed in {total_time:.2f}s")
    
    # Process results
    successful = 0
    failed = 0
    
    for response in responses:
        if isinstance(response, Exception):
            failed += 1
            print(f"Failed request: {response}")
        else:
            successful += 1
            print(f"✓ {response['url']} - {response['latency']:.2f}s")
    
    print(f"\nSummary: {successful} successful, {failed} failed")
    print(f"Concurrent requests are {len(urls)/total_time:.1f}x faster than sequential!")

# Run the HTTP simulation
asyncio.run(fetch_multiple_urls())
````

````python
# Async file operations
print("\n=== ASYNC FILE OPERATIONS ===")

import aiofiles
import os

# Check if aiofiles is available
try:
    import aiofiles
    AIOFILES_AVAILABLE = True
except ImportError:
    AIOFILES_AVAILABLE = False
    # Dependency missing? Install once with: pip install -r requirements.txt
    import aiofiles
    AIOFILES_AVAILABLE = True

async def async_file_operations():
    """Demonstrate async file operations."""
    
    # Create sample data
    sample_data = """Async file operations allow you to perform I/O operations
without blocking the event loop. This is particularly useful
when dealing with multiple files or network operations.

Key benefits:
- Non-blocking I/O
- Better resource utilization
- Improved application responsiveness
- Scalable concurrent operations
"""
    
    # Write multiple files concurrently
    async def write_file(filename, content):
        async with aiofiles.open(filename, 'w') as f:
            await f.write(content)
        print(f"Wrote {len(content)} chars to {filename}")
    
    # Create multiple files
    files_to_create = {
        'async_file1.txt': sample_data,
        'async_file2.txt': sample_data.replace('Async', 'Concurrent'),
        'async_file3.txt': sample_data.replace('Async', 'Parallel'),
    }
    
    print("Creating files concurrently...")
    await asyncio.gather(*[
        write_file(filename, content)
        for filename, content in files_to_create.items()
    ])
    
    # Read files concurrently
    async def read_file(filename):
        async with aiofiles.open(filename, 'r') as f:
            content = await f.read()
        return filename, len(content), content[:50] + "..."
    
    print("\nReading files concurrently...")
    results = await asyncio.gather(*[
        read_file(filename) for filename in files_to_create.keys()
    ])
    
    for filename, size, preview in results:
        print(f"{filename}: {size} chars - {preview}")
    
    # Clean up
    for filename in files_to_create.keys():
        os.remove(filename)
    
    print("\nCleaned up demo files")

if AIOFILES_AVAILABLE:
    asyncio.run(async_file_operations())
else:
    print("aiofiles not available for file operations demo")
````

````python
# Error handling in async code
print("\n=== ERROR HANDLING IN ASYNC CODE ===")

async def unreliable_operation(name, success_rate=0.7):
    """An operation that sometimes fails."""
    await asyncio.sleep(0.1)
    
    if random.random() < success_rate:
        return f"{name} succeeded"
    else:
        raise Exception(f"{name} failed")

async def error_handling_demo():
    """Demonstrate error handling in async code."""
    
    print("=== Basic Error Handling ===")
    try:
        result = await unreliable_operation("Operation 1")
        print(f"✓ {result}")
    except Exception as e:
        print(f"✗ {e}")
    
    print("\n=== Concurrent Operations with Error Handling ===")
    
    # Run multiple operations concurrently with error handling
    operations = [f"Op {i}" for i in range(1, 6)]
    
    # Method 1: Handle exceptions individually
    async def safe_operation(name):
        try:
            return await unreliable_operation(name, 0.6)
        except Exception as e:
            return f"ERROR: {e}"
    
    print("Individual error handling:")
    results = await asyncio.gather(*[safe_operation(op) for op in operations])
    for result in results:
        print(f"  {result}")
    
    # Method 2: Let exceptions propagate and handle at gather level
    print("\nGather with return_exceptions:")
    results = await asyncio.gather(*[
        unreliable_operation(op, 0.4) for op in operations
    ], return_exceptions=True)
    
    successful = 0
    failed = 0
    
    for i, result in enumerate(results):
        if isinstance(result, Exception):
            print(f"  ✗ {operations[i]}: {result}")
            failed += 1
        else:
            print(f"  ✓ {operations[i]}: {result}")
            successful += 1
    
    print(f"\nSummary: {successful} successful, {failed} failed")

# Run error handling demo
asyncio.run(error_handling_demo())
````
