## 🏭 Creational Patterns

Creational patterns deal with object creation mechanisms, trying to create objects in a manner suitable to the situation.

````python
# Singleton Pattern - Ensure only one instance exists
print("=== SINGLETON PATTERN ===")

class Singleton:
    """Classic Singleton implementation."""
    _instance = None
    
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
    
    def __init__(self):
        if not hasattr(self, '_initialized'):
            self._initialized = True
            self.data = "Singleton data"

# Test singleton
s1 = Singleton()
s1.data = "Modified by s1"

s2 = Singleton()
print(f"s1 is s2: {s1 is s2}")
print(f"s2.data: {s2.data}")
print(f"Same instance: {id(s1) == id(s2)}")

# Database connection singleton
class DatabaseConnection:
    """Database connection singleton."""
    _instance = None
    
    def __new__(cls, host="localhost", port=5432):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
    
    def __init__(self, host="localhost", port=5432):
        if not hasattr(self, '_initialized'):
            self._initialized = True
            self.host = host
            self.port = port
            print(f"Connected to database at {host}:{port}")
    
    def query(self, sql):
        return f"Executed: {sql}"

# Test database singleton
db1 = DatabaseConnection("prod-db", 5432)
db2 = DatabaseConnection("dev-db", 5432)  # Ignored

print(f"db1 is db2: {db1 is db2}")
print(f"Connection: {db1.host}:{db1.port}")
print(db1.query("SELECT * FROM users"))
````

````python
# Factory Pattern - Create objects without specifying exact classes
print("\n=== FACTORY PATTERN ===")

# Product interface
class Shape:
    """Abstract shape class."""
    def draw(self):
        raise NotImplementedError
    
    def area(self):
        raise NotImplementedError

# Concrete products
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def draw(self):
        return f"Drawing circle with radius {self.radius}"
    
    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def draw(self):
        return f"Drawing rectangle {self.width}x{self.height}"
    
    def area(self):
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def draw(self):
        return f"Drawing triangle with base {self.base}, height {self.height}"
    
    def area(self):
        return 0.5 * self.base * self.height

# Factory class
class ShapeFactory:
    """Factory for creating shapes."""
    
    @staticmethod
    def create_shape(shape_type, **kwargs):
        """Create a shape based on type and parameters."""
        shapes = {
            'circle': Circle,
            'rectangle': Rectangle,
            'triangle': Triangle
        }
        
        shape_class = shapes.get(shape_type.lower())
        if not shape_class:
            raise ValueError(f"Unknown shape type: {shape_type}")
        
        return shape_class(**kwargs)

# Test factory pattern
shapes = [
    ShapeFactory.create_shape('circle', radius=5),
    ShapeFactory.create_shape('rectangle', width=4, height=6),
    ShapeFactory.create_shape('triangle', base=3, height=4)
]

for shape in shapes:
    print(f"{shape.draw()} - Area: {shape.area():.2f}")

# Factory method pattern
class GUIFactory:
    """Abstract factory for GUI components."""
    
    def create_button(self):
        raise NotImplementedError
    
    def create_checkbox(self):
        raise NotImplementedError

class WindowsFactory(GUIFactory):
    def create_button(self):
        return "Windows Button"
    
    def create_checkbox(self):
        return "Windows Checkbox"

class MacFactory(GUIFactory):
    def create_button(self):
        return "Mac Button"
    
    def create_checkbox(self):
        return "Mac Checkbox"

# Client code
def create_ui(factory):
    button = factory.create_button()
    checkbox = factory.create_checkbox()
    return f"UI: {button}, {checkbox}"

windows_ui = create_ui(WindowsFactory())
mac_ui = create_ui(MacFactory())

print(f"\n{windows_ui}")
print(f"{mac_ui}")
````

````python
# Builder Pattern - Construct complex objects step by step
print("\n=== BUILDER PATTERN ===")

class Pizza:
    """Pizza class with many optional components."""
    
    def __init__(self):
        self.dough = None
        self.sauce = None
        self.cheese = None
        self.toppings = []
    
    def __str__(self):
        toppings_str = ", ".join(self.toppings) if self.toppings else "no toppings"
        return f"Pizza: {self.dough} dough, {self.sauce} sauce, {self.cheese} cheese, {toppings_str}"

class PizzaBuilder:
    """Builder for creating pizzas step by step."""
    
    def __init__(self):
        self.pizza = Pizza()
    
    def set_dough(self, dough):
        self.pizza.dough = dough
        return self
    
    def set_sauce(self, sauce):
        self.pizza.sauce = sauce
        return self
    
    def set_cheese(self, cheese):
        self.pizza.cheese = cheese
        return self
    
    def add_topping(self, topping):
        self.pizza.toppings.append(topping)
        return self
    
    def build(self):
        return self.pizza

# Fluent interface
margherita = (PizzaBuilder()
              .set_dough("thin")
              .set_sauce("tomato")
              .set_cheese("mozzarella")
              .add_topping("basil")
              .build())

pepperoni = (PizzaBuilder()
             .set_dough("thick")
             .set_sauce("tomato")
             .set_cheese("mozzarella")
             .add_topping("pepperoni")
             .add_topping("mushrooms")
             .build())

print(margherita)
print(pepperoni)

# Director pattern (optional)
class PizzaDirector:
    """Director for creating predefined pizzas."""
    
    @staticmethod
    def make_margherita():
        return (PizzaBuilder()
                .set_dough("thin")
                .set_sauce("tomato")
                .set_cheese("mozzarella")
                .add_topping("basil")
                .build())
    
    @staticmethod
    def make_pepperoni():
        return (PizzaBuilder()
                .set_dough("thick")
                .set_sauce("tomato")
                .set_cheese("mozzarella")
                .add_topping("pepperoni")
                .build())

veggie = PizzaDirector.make_margherita()
veggie.toppings.append("olives")
veggie.toppings.append("peppers")

print(f"\nDirector-created pizzas:")
print(f"Margherita: {PizzaDirector.make_margherita()}")
print(f"Pepperoni: {PizzaDirector.make_pepperoni()}")
print(f"Custom veggie: {veggie}")
````

## 🏗️ Structural Patterns

Structural patterns deal with object composition and typically use inheritance to compose interfaces.

````python
# Adapter Pattern - Convert interface of a class into another interface
print("=== ADAPTER PATTERN ===")

# Target interface
class MediaPlayer:
    """Interface for media players."""
    def play(self, audio_type, filename):
        raise NotImplementedError

# Adaptee classes (existing interfaces)
class MP3Player:
    def play_mp3(self, filename):
        return f"Playing MP3: {filename}"

class VLCPlayer:
    def play_vlc(self, filename):
        return f"Playing VLC: {filename}"
    
    def play_mp4(self, filename):
        return f"Playing MP4: {filename}"

# Adapter classes
class MP3Adapter(MediaPlayer):
    """Adapter for MP3 player."""
    
    def __init__(self, mp3_player):
        self.mp3_player = mp3_player
    
    def play(self, audio_type, filename):
        if audio_type == "mp3":
            return self.mp3_player.play_mp3(filename)
        else:
            return f"Unsupported format: {audio_type}"

class VLCAdapter(MediaPlayer):
    """Adapter for VLC player."""
    
    def __init__(self, vlc_player):
        self.vlc_player = vlc_player
    
    def play(self, audio_type, filename):
        if audio_type == "vlc":
            return self.vlc_player.play_vlc(filename)
        elif audio_type == "mp4":
            return self.vlc_player.play_mp4(filename)
        else:
            return f"Unsupported format: {audio_type}"

# Audio player using adapters
class AudioPlayer:
    """Audio player that uses adapters."""
    
    def __init__(self):
        self.mp3_adapter = MP3Adapter(MP3Player())
        self.vlc_adapter = VLCAdapter(VLCPlayer())
    
    def play(self, audio_type, filename):
        if audio_type == "mp3":
            return self.mp3_adapter.play(audio_type, filename)
        elif audio_type in ["vlc", "mp4"]:
            return self.vlc_adapter.play(audio_type, filename)
        else:
            return f"Unsupported audio type: {audio_type}"

# Test adapter pattern
player = AudioPlayer()

formats_to_test = [
    ("mp3", "song.mp3"),
    ("vlc", "movie.vlc"),
    ("mp4", "video.mp4"),
    ("wav", "audio.wav")  # Unsupported
]

for audio_type, filename in formats_to_test:
    result = player.play(audio_type, filename)
    print(result)
````

````python
# Decorator Pattern - Add behavior to objects dynamically
print("\n=== DECORATOR PATTERN ===")

# Component interface
class Coffee:
    """Base coffee class."""
    def cost(self):
        raise NotImplementedError
    
    def description(self):
        raise NotImplementedError

# Concrete component
class SimpleCoffee(Coffee):
    def cost(self):
        return 2.0
    
    def description(self):
        return "Simple coffee"

# Base decorator
class CoffeeDecorator(Coffee):
    """Base decorator class."""
    
    def __init__(self, coffee):
        self._coffee = coffee
    
    def cost(self):
        return self._coffee.cost()
    
    def description(self):
        return self._coffee.description()

# Concrete decorators
class MilkDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 0.5
    
    def description(self):
        return self._coffee.description() + ", milk"

class SugarDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 0.2
    
    def description(self):
        return self._coffee.description() + ", sugar"

class WhipDecorator(CoffeeDecorator):
    def cost(self):
        return self._coffee.cost() + 0.7
    
    def description(self):
        return self._coffee.description() + ", whip cream"

# Test decorator pattern
coffee = SimpleCoffee()
print(f"Base coffee: ${coffee.cost():.2f} - {coffee.description()}")

# Add decorators
coffee_with_milk = MilkDecorator(coffee)
print(f"With milk: ${coffee_with_milk.cost():.2f} - {coffee_with_milk.description()}")

fancy_coffee = WhipDecorator(SugarDecorator(MilkDecorator(SimpleCoffee())))
print(f"Fancy coffee: ${fancy_coffee.cost():.2f} - {fancy_coffee.description()}")

# Python's built-in decorator syntax (function decorators)
def logging_decorator(func):
    """Function decorator that adds logging."""
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@logging_decorator
def add_numbers(a, b):
    """Add two numbers."""
    return a + b

@logging_decorator  
def multiply_numbers(a, b):
    """Multiply two numbers."""
    return a * b

print(f"\nFunction decorators:")
result1 = add_numbers(3, 5)
result2 = multiply_numbers(4, 6)

# Multiple decorators
def timing_decorator(func):
    import time
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {(end-start)*1000:.2f}ms")
        return result
    return wrapper

@timing_decorator
@logging_decorator
def slow_operation(n):
    """A slow operation for demonstration."""
    import time
    time.sleep(0.1)
    return sum(range(n))

print(f"\nMultiple decorators:")
result = slow_operation(1000)
````

````python
# Facade Pattern - Provide a unified interface to a set of interfaces
print("\n=== FACADE PATTERN ===")

# Complex subsystems
class CPU:
    def freeze(self):
        print("CPU: Freezing...")
    
    def jump(self, position):
        print(f"CPU: Jumping to {position}")
    
    def execute(self):
        print("CPU: Executing...")

class Memory:
    def load(self, position, data):
        print(f"Memory: Loading {data} at {position}")

class HardDrive:
    def read(self, lba, size):
        print(f"HardDrive: Reading {size} bytes from LBA {lba}")
        return f"Data from sector {lba}"

# Facade class
class ComputerFacade:
    """Facade that simplifies computer startup."""
    
    def __init__(self):
        self.cpu = CPU()
        self.memory = Memory()
        self.hard_drive = HardDrive()
    
    def start_computer(self):
        """Simplified interface for starting computer."""
        print("Starting computer...")
        self.cpu.freeze()
        
        # Load boot loader
        boot_data = self.hard_drive.read(0, 1024)
        self.memory.load(0, boot_data)
        
        self.cpu.jump(0)
        self.cpu.execute()
        
        print("Computer started successfully!")

# Test facade pattern
computer = ComputerFacade()
computer.start_computer()

# Without facade (complex)
print("\nWithout facade (complex):")
cpu = CPU()
memory = Memory()
hd = HardDrive()

cpu.freeze()
boot_data = hd.read(0, 1024)
memory.load(0, boot_data)
cpu.jump(0)
cpu.execute()

# Web framework facade example
class WebFrameworkFacade:
    """Facade for web framework operations."""
    
    def __init__(self):
        # In a real framework, these would be complex subsystems
        pass
    
    def create_app(self, name):
        """Create a new web application."""
        print(f"Setting up application: {name}")
        print("- Configuring routes")
        print("- Setting up middleware")
        print("- Initializing database")
        print("- Starting server")
        return f"{name} application ready"
    
    def deploy_app(self, app_name):
        """Deploy application to production."""
        print(f"Deploying {app_name}...")
        print("- Building application")
        print("- Running tests")
        print("- Creating Docker image")
        print("- Deploying to server")
        return f"{app_name} deployed successfully"

# Test web framework facade
framework = WebFrameworkFacade()
app = framework.create_app("MyWebApp")
print(f"\n{app}")

deployment = framework.deploy_app("MyWebApp")
print(deployment)
````

## 🎭 Behavioral Patterns

Behavioral patterns deal with communication between objects and how responsibilities are distributed.

````python
# Observer Pattern - Define a one-to-many dependency between objects
print("=== OBSERVER PATTERN ===")

# Observer interface
class Observer:
    """Observer interface."""
    def update(self, subject):
        raise NotImplementedError

# Subject interface
class Subject:
    """Subject interface."""
    
    def __init__(self):
        self._observers = []
    
    def attach(self, observer):
        self._observers.append(observer)
    
    def detach(self, observer):
        self._observers.remove(observer)
    
    def notify(self):
        for observer in self._observers:
            observer.update(self)

# Concrete subject
class WeatherStation(Subject):
    """Weather station that notifies observers of changes."""
    
    def __init__(self):
        super().__init__()
        self._temperature = 0
        self._humidity = 0
    
    @property
    def temperature(self):
        return self._temperature
    
    @temperature.setter
    def temperature(self, value):
        self._temperature = value
        self.notify()
    
    @property
    def humidity(self):
        return self._humidity
    
    @humidity.setter
    def humidity(self, value):
        self._humidity = value
        self.notify()

# Concrete observers
class Display(Observer):
    """Display that shows current weather."""
    
    def __init__(self, name):
        self.name = name
    
    def update(self, subject):
        if isinstance(subject, WeatherStation):
            print(f"{self.name}: Temperature = {subject.temperature}°C, "
                  f"Humidity = {subject.humidity}%")

class Logger(Observer):
    """Logger that records weather changes."""
    
    def update(self, subject):
        if isinstance(subject, WeatherStation):
            print(f"[LOG] Weather changed: T={subject.temperature}°C, "
                  f"H={subject.humidity}%")

class AlertSystem(Observer):
    """Alert system that triggers warnings."""
    
    def update(self, subject):
        if isinstance(subject, WeatherStation):
            if subject.temperature > 30:
                print("🚨 ALERT: High temperature detected!")
            if subject.humidity > 80:
                print("🚨 ALERT: High humidity detected!")

# Test observer pattern
station = WeatherStation()

# Attach observers
display1 = Display("Main Display")
display2 = Display("Backup Display")
logger = Logger()
alert = AlertSystem()

station.attach(display1)
station.attach(display2)
station.attach(logger)
station.attach(alert)

# Change weather conditions
print("Setting initial conditions:")
station.temperature = 25
station.humidity = 60

print("\nChanging temperature:")
station.temperature = 35  # Should trigger alert

print("\nChanging humidity:")
station.humidity = 85  # Should trigger alert

# Detach an observer
station.detach(display2)
print("\nAfter detaching backup display:")
station.temperature = 20
````

````python
# Strategy Pattern - Define a family of algorithms, encapsulate each one, and make them interchangeable
print("\n=== STRATEGY PATTERN ===")

# Strategy interface
class PaymentStrategy:
    """Payment strategy interface."""
    def pay(self, amount):
        raise NotImplementedError

# Concrete strategies
class CreditCardPayment(PaymentStrategy):
    def __init__(self, card_number, expiry, cvv):
        self.card_number = card_number
        self.expiry = expiry
        self.cvv = cvv
    
    def pay(self, amount):
        last_four = self.card_number[-4:]
        return f"Paid ${amount:.2f} using credit card ending in {last_four}"

class PayPalPayment(PaymentStrategy):
    def __init__(self, email):
        self.email = email
    
    def pay(self, amount):
        return f"Paid ${amount:.2f} using PayPal account {self.email}"

class BitcoinPayment(PaymentStrategy):
    def __init__(self, wallet_address):
        self.wallet_address = wallet_address
    
    def pay(self, amount):
        short_addr = self.wallet_address[:8] + "..."
        return f"Paid ${amount:.2f} in Bitcoin to wallet {short_addr}"

# Context class
class ShoppingCart:
    """Shopping cart that uses different payment strategies."""
    
    def __init__(self):
        self.items = []
        self._payment_strategy = None
    
    def add_item(self, item, price):
        self.items.append((item, price))
    
    def set_payment_strategy(self, strategy):
        self._payment_strategy = strategy
    
    def checkout(self):
        if not self._payment_strategy:
            return "No payment method selected"
        
        total = sum(price for _, price in self.items)
        return self._payment_strategy.pay(total)

# Test strategy pattern
cart = ShoppingCart()
cart.add_item("Laptop", 999.99)
cart.add_item("Mouse", 29.99)
cart.add_item("Keyboard", 79.99)

# Try different payment strategies
credit_card = CreditCardPayment("1234567890123456", "12/25", "123")
cart.set_payment_strategy(credit_card)
print(f"Credit Card: {cart.checkout()}")

paypal = PayPalPayment("user@example.com")
cart.set_payment_strategy(paypal)
print(f"PayPal: {cart.checkout()}")

bitcoin = BitcoinPayment("1A2B3C4D5E6F7890ABCDEF")
cart.set_payment_strategy(bitcoin)
print(f"Bitcoin: {cart.checkout()}")

# Strategy with sorting algorithms
from typing import List

class SortStrategy:
    """Sorting strategy interface."""
    def sort(self, data: List[int]) -> List[int]:
        raise NotImplementedError

class BubbleSort(SortStrategy):
    def sort(self, data: List[int]) -> List[int]:
        arr = data.copy()
        n = len(arr)
        for i in range(n):
            for j in range(0, n-i-1):
                if arr[j] > arr[j+1]:
                    arr[j], arr[j+1] = arr[j+1], arr[j]
        return arr

class QuickSort(SortStrategy):
    def sort(self, data: List[int]) -> List[int]:
        arr = data.copy()
        if len(arr) <= 1:
            return arr
        
        pivot = arr[len(arr) // 2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        
        return self.sort(left) + middle + self.sort(right)

class Sorter:
    """Sorter that uses different sorting strategies."""
    
    def __init__(self, strategy: SortStrategy):
        self._strategy = strategy
    
    def set_strategy(self, strategy: SortStrategy):
        self._strategy = strategy
    
    def sort(self, data: List[int]) -> List[int]:
        return self._strategy.sort(data)

# Test sorting strategies
data = [64, 34, 25, 12, 22, 11, 90]
sorter = Sorter(BubbleSort())
print(f"\nOriginal data: {data}")
print(f"Bubble sort: {sorter.sort(data)}")

sorter.set_strategy(QuickSort())
print(f"Quick sort: {sorter.sort(data)}")
````

````python
# Command Pattern - Encapsulate a request as an object
print("\n=== COMMAND PATTERN ===")

# Command interface
class Command:
    """Command interface."""
    def execute(self):
        raise NotImplementedError
    
    def undo(self):
        raise NotImplementedError

# Receiver classes
class Light:
    """Light that can be turned on/off."""
    
    def __init__(self, location):
        self.location = location
        self.is_on = False
    
    def turn_on(self):
        self.is_on = True
        print(f"{self.location} light turned on")
    
    def turn_off(self):
        self.is_on = False
        print(f"{self.location} light turned off")

class Stereo:
    """Stereo system."""
    
    def __init__(self, location):
        self.location = location
        self.is_on = False
        self.volume = 0
    
    def turn_on(self):
        self.is_on = True
        print(f"{self.location} stereo turned on")
    
    def turn_off(self):
        self.is_on = False
        print(f"{self.location} stereo turned off")
    
    def set_volume(self, volume):
        self.volume = volume
        print(f"{self.location} stereo volume set to {volume}")

# Concrete commands
class LightOnCommand(Command):
    """Command to turn on a light."""
    
    def __init__(self, light):
        self.light = light
    
    def execute(self):
        self.light.turn_on()
    
    def undo(self):
        self.light.turn_off()

class LightOffCommand(Command):
    """Command to turn off a light."""
    
    def __init__(self, light):
        self.light = light
    
    def execute(self):
        self.light.turn_off()
    
    def undo(self):
        self.light.turn_on()

class StereoOnWithVolumeCommand(Command):
    """Command to turn on stereo and set volume."""
    
    def __init__(self, stereo):
        self.stereo = stereo
    
    def execute(self):
        self.stereo.turn_on()
        self.stereo.set_volume(11)
    
    def undo(self):
        self.stereo.turn_off()

# Invoker (remote control)
class RemoteControl:
    """Remote control that executes commands."""
    
    def __init__(self):
        self.commands = {}
        self.undo_stack = []
    
    def set_command(self, slot, command):
        self.commands[slot] = command
    
    def press_button(self, slot):
        if slot in self.commands:
            command = self.commands[slot]
            command.execute()
            self.undo_stack.append(command)
            return f"Executed command in slot {slot}"
        return f"No command in slot {slot}"
    
    def press_undo(self):
        if self.undo_stack:
            command = self.undo_stack.pop()
            command.undo()
            return "Undid last command"
        return "Nothing to undo"

# Test command pattern
living_room_light = Light("Living Room")
kitchen_light = Light("Kitchen")
stereo = Stereo("Living Room")

light_on = LightOnCommand(living_room_light)
light_off = LightOffCommand(living_room_light)
kitchen_on = LightOnCommand(kitchen_light)
stereo_on = StereoOnWithVolumeCommand(stereo)

remote = RemoteControl()
remote.set_command("living_on", light_on)
remote.set_command("living_off", light_off)
remote.set_command("kitchen_on", kitchen_on)
remote.set_command("stereo_on", stereo_on)

# Use remote control
print("Using remote control:")
print(remote.press_button("living_on"))
print(remote.press_button("stereo_on"))
print(remote.press_button("kitchen_on"))

print("\nUndoing commands:")
print(remote.press_undo())
print(remote.press_undo())

# Macro command (composite command)
class MacroCommand(Command):
    """Command that executes multiple commands."""
    
    def __init__(self, commands):
        self.commands = commands
    
    def execute(self):
        for command in self.commands:
            command.execute()
    
    def undo(self):
        # Undo in reverse order
        for command in reversed(self.commands):
            command.undo()

party_mode = MacroCommand([light_on, stereo_on])
remote.set_command("party", party_mode)

print(f"\nMacro command: {remote.press_button('party')}")
print(f"Undo macro: {remote.press_undo()}")
````
